```python
rng = np.random.default_rng()
arr = np.arange(10)
arr
```

```python
rng.shuffle(arr)
arr
```

```python
arr = np.arange(9).reshape((3, 3))
arr
```

```python
rng.shuffle(arr)
arr
```

```python
arr = np.arange(9).reshape((3, 3))
arr
```

```python
rng.shuffle(arr, axis=1)
arr
```
