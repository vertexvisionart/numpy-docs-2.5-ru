```python
import numpy as np
```

Применяем метод трапеций к равномерно расположенным точкам:


```python
np.trapezoid([1, 2, 3])
```

Расстояния между точками можно задать аргументами ``x`` или ``dx``:


```python
np.trapezoid([1, 2, 3], x=[4, 6, 8])
```

```python
np.trapezoid([1, 2, 3], dx=2)
```

Убывающий ``x`` соответствует интегрированию в обратном направлении:


```python
np.trapezoid([1, 2, 3], x=[8, 6, 4])
```

В более общем случае ``x`` используется для интегрирования вдоль параметрической кривой. Оценим интеграл $\int_0^1 x^2 = 1/3$ следующим образом:


```python
x = np.linspace(0, 1, num=50)
y = x**2
np.trapezoid(y, x)
```

Или оцените площадь круга; обратите внимание, что мы повторяем точку выборки, замыкающую кривую:

```python
theta = np.linspace(0, 2 * np.pi, num=1000, endpoint=True)
np.trapezoid(np.cos(theta), x=np.sin(theta))
```

``np.trapezoid`` можно применять вдоль заданной оси, выполняя несколько вычислений одним вызовом:


```python
a = np.arange(6).reshape(2, 3)
a
```

```python
np.trapezoid(a, axis=0)
```

```python
np.trapezoid(a, axis=1)
```
