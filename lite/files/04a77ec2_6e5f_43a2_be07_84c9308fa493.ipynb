Тип данных второго аргумента может изменить тип результата и в некоторых случаях привести к неожиданным результатам. См. раздел «Правила приведения типов»:


```python
a = np.left_shift(np.uint8(255), np.int64(1))  # Expect 254
print(a, type(a)) # Unexpected result due to upcasting
```

**Комментарии к примеру на русском**

Ожидается 254

Неожиданный результат из-за приведения к более широкому типу

```python
b = np.left_shift(np.uint8(255), np.uint8(1))
print(b, type(b))
```
