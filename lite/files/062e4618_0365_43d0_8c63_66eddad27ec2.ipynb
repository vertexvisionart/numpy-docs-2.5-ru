```python
import numpy as np
dt = np.dtype(float, metadata={"key": "value"})
dt.metadata["key"]
```

```python
arr = np.array([1, 2, 3], dtype=dt)
arr.dtype.metadata
```

Сейчас сложение массивов с одинаковыми типами данных сохраняет метаданные:

```python
(arr + arr).dtype.metadata
```

Если массивы имеют разные метаданные dtype, выигрывает первый:

```python
dt2 = np.dtype(float, metadata={"key2": "value2"})
arr2 = np.array([3, 2, 1], dtype=dt2)
print((arr + arr2).dtype.metadata)
```
