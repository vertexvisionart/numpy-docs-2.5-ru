Сгенерируем выборку из распределения:


```python
ngood, nbad, nsamp = 100, 2, 10
```

```python
s = np.random.hypergeometric(ngood, nbad, nsamp, 1000)
from matplotlib.pyplot import hist
hist(s)
```

Предположим, у вас есть урна с 15 белыми и 15 черными шариками. Если вы наугад вытащите 15 шариков, насколько вероятно, что 12 или более из них одного цвета?

```python
s = np.random.hypergeometric(15, 15, 15, 100000)
sum(s>=12)/100000. + sum(s<=3)/100000.
```
