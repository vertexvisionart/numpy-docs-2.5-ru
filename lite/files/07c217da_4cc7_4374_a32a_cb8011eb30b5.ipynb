Основные примеры

```python
import numpy as np
np.can_cast(np.int32, np.int64)
```

```python
np.can_cast(np.float64, complex)
```

```python
np.can_cast(complex, float)
```

```python
np.can_cast('i8', 'f8')
```

```python
np.can_cast('i8', 'f4')
```

```python
np.can_cast('i4', 'S4')
```
