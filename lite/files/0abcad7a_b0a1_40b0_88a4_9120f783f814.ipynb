```python
import numpy as np
x = np.array([[1,2,3]])
y = np.array([[4],[5]])
np.broadcast_arrays(x, y)
```

Полезный приём для получения непрерывно расположенных в памяти копий вместо представлений с разрывами:


```python
[np.array(a) for a in np.broadcast_arrays(x, y)]
```
