Пример из Glantz[1], с. 47–40:

Рассматриваются две группы: дети родителей с диабетом (25 человек)
и дети родителей без диабета (контрольная группа, 25 человек).
Измеряли уровень глюкозы в крови натощак. В основной группе среднее
значение составило 86.1, в контрольной — 82.2. Стандартные отклонения
равнялись 2.09 и 2.49 соответственно. Согласуются ли эти данные с нулевой
гипотезой о том, что наличие диабета у родителей не влияет на уровень
глюкозы в крови их детей? Расчёт F-статистики по данным даёт 36.01.

Генерируем выборку из распределения:


```python
dfnum = 1. # between group degrees of freedom
dfden = 48. # within groups degrees of freedom
s = np.random.f(dfnum, dfden, 1000)
```

Нижняя граница для верхнего 1% значений выборки:


```python
np.sort(s)[-10]
```

Таким образом, вероятность того, что F-статистика превысит 7,62, составляет около 1%. Наблюдаемое значение равно 36, поэтому нулевая гипотеза отвергается на уровне значимости 1%.
