Для двумерных массивов это матричное произведение:


```python
import numpy as np
a = np.array([[1, 0],
              [0, 1]])
b = np.array([[4, 1],
              [2, 2]])
np.matmul(a, b)
```

При сочетании двумерного и одномерного массивов результат обычный.


```python
a = np.array([[1, 0],
              [0, 1]])
b = np.array([1, 2])
np.matmul(a, b)
```

```python
np.matmul(b, a)
```

Для наборов массивов действуют обычные правила трансляции массивов.


```python
a = np.arange(2 * 2 * 4).reshape((2, 2, 4))
b = np.arange(2 * 2 * 4).reshape((2, 4, 2))
np.matmul(a,b).shape
```

```python
np.matmul(a, b)[0, 1, 1]
```

```python
sum(a[0, 1, :] * b[0 , :, 1])
```

Для двух векторов возвращается скалярное произведение, но комплексное сопряжение не применяется ни к одному из аргументов:


```python
np.matmul([2j, 3j], [2j, 3j])
```

Умножение на скаляр вызывает ошибку.


```python
np.matmul([1,2], 3)
```

Для массивов ndarray оператор ``@`` служит краткой записью ``np.matmul``.


```python
x1 = np.array([2j, 3j])
x2 = np.array([2j, 3j])
x1 @ x2
```
