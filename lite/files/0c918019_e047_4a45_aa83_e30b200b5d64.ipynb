В примере из книги Далгаарда, с. 83 [1], суточное потребление энергии
11 женщинами в килоджоулях (кДж) составляет:


```python
intake = np.array([5260., 5470, 5640, 6180, 6390, 6515, 6805, 7515, \
                   7515, 8230, 8770])
```

Отличается ли их потребление энергии систематически от рекомендованного
значения 7725 кДж? Нулевая гипотеза — отсутствие отклонения,
альтернативная — наличие эффекта любого знака,
поэтому критерий будет двусторонним.

Поскольку мы оцениваем среднее и имеем в выборке N=11 значений,
число степеней свободы равно N-1=10. Выбираем доверительный уровень 95%
и вычисляем t-статистику по выборочному среднему и выборочному стандартному
отклонению потребления энергии. Используем ddof=1, чтобы при расчёте
выборочного стандартного отклонения исходить из несмещённой оценки дисперсии.
Итоговая оценка стандартного отклонения всё же смещена из-за вогнутости
функции квадратного корня.

Примечание переводчика: в оригинале 95% ошибочно названо уровнем значимости;
соответствующий уровень значимости равен 5%.


```python
np.mean(intake)
```

```python
intake.std(ddof=1)
```

```python
t = (np.mean(intake)-7725)/(intake.std(ddof=1)/np.sqrt(len(intake)))
t
```

Сгенерируем 1000000 значений из t-распределения Стьюдента с подходящим числом степеней свободы.


```python
import matplotlib.pyplot as plt
s = np.random.standard_t(10, size=1000000)
h = plt.hist(s, bins=100, density=True)
```

Попадает ли наша t-статистика в одну из двух критических областей в хвостах распределения?

```python
np.sum(np.abs(t) < np.abs(s)) / float(len(s))
```

p-значение двустороннего теста составляет около 1,83%, что ниже заранее выбранного уровня значимости 5%.

Следовательно, при истинной нулевой гипотезе вероятность наблюдать столь же крайние значения потребления слишком мала, и нулевая гипотеза об отсутствии отклонения отвергается.
