````python
import numpy as np
````

``mask`` можно инициализировать массивом логических значений той же формы, что и ``data``.


````python
data = np.arange(6).reshape((2, 3))
np.ma.MaskedArray(data, mask=[[False, True, False],
                              [False, False, True]])
````

Также ``mask`` можно инициализировать логическим массивом, заполненным одним значением и имеющим ту же форму, что и ``data``: для этого передайте скалярное логическое значение:


````python
np.ma.MaskedArray(data, mask=False)
````

````python
np.ma.MaskedArray(data, mask=True)
````

**Примечание.** Если маску `mask` инициализируют скалярным булевым значением, рекомендуется использовать `True` или `False`, а не `np.True_` или `np.False_`. Причина: атрибут `nomask` внутренне представлен объектом `np.False_`.

```python
>>> np.False_ is np.ma.nomask
True
```
