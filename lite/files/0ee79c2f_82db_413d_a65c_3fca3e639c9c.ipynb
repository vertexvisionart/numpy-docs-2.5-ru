```python
import numpy as np
nx, ny = (3, 2)
x = np.linspace(0, 1, nx)
y = np.linspace(0, 1, ny)
xv, yv = np.meshgrid(x, y)
xv
```

```python
yv
```

Результатом `meshgrid` является координатная сетка:

```python
import matplotlib.pyplot as plt
plt.plot(xv, yv, marker='o', color='k', linestyle='none')
plt.show()
```

Для экономии памяти и времени вычислений можно создавать разреженные выходные массивы.


```python
xv, yv = np.meshgrid(x, y, sparse=True)
xv
```

```python
yv
```

`meshgrid` очень удобна для вычисления значений функций на сетке. Если функция зависит от всех координат, можно использовать как плотные, так и разреженные выходные массивы.

```python
x = np.linspace(-5, 5, 101)
y = np.linspace(-5, 5, 101)
# full coordinate arrays
xx, yy = np.meshgrid(x, y)
zz = np.sqrt(xx**2 + yy**2)
xx.shape, yy.shape, zz.shape
```

**Комментарии к примеру на русском**

Полные координатные массивы

```python
# sparse coordinate arrays
xs, ys = np.meshgrid(x, y, sparse=True)
zs = np.sqrt(xs**2 + ys**2)
xs.shape, ys.shape, zs.shape
```

**Комментарии к примеру на русском**

Разреженные координатные массивы

```python
np.array_equal(zz, zs)
```

```python
h = plt.contourf(x, y, zs)
plt.axis('scaled')
plt.colorbar()
plt.show()
```
