Диагонали, прилегающие к основной диагонали, можно получить с помощью аргумента `offset`:

```python
a = np.arange(9).reshape(3, 3)
a
```

```python
np.linalg.diagonal(a, offset=1)  # First superdiagonal
```

**Комментарии к примеру на русском**

Первая диагональ над главной

```python
np.linalg.diagonal(a, offset=2)  # Second superdiagonal
```

**Комментарии к примеру на русском**

Вторая диагональ над главной

```python
np.linalg.diagonal(a, offset=-1)  # First subdiagonal
```

**Комментарии к примеру на русском**

Первая диагональ под главной

```python
np.linalg.diagonal(a, offset=-2)  # Second subdiagonal
```

**Комментарии к примеру на русском**

Вторая диагональ под главной

Антидиагональ может быть получена путем изменения порядка элементов с использованием либо `numpy.flipud`, либо `numpy.fliplr`.

```python
a = np.arange(9).reshape(3, 3)
a
```

```python
np.linalg.diagonal(np.fliplr(a))  # Horizontal flip
```

**Комментарии к примеру на русском**

Отражение по горизонтали

```python
np.linalg.diagonal(np.flipud(a))  # Vertical flip
```

**Комментарии к примеру на русском**

Отражение по вертикали

Порядок извлечения диагонали зависит от используемой функции отражения.
