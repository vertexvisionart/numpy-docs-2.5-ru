```python
import numpy as np
```

```python
a = np.datetime64("0000-01-01", "us")
b = np.datetime64("1600-01-01", "us")
b - a
```

Вычисленный результат, `50491123200` секунд, получается умножением числа прошедших дней (`584388`) на `86400` секунд. Это число секунд по часам, синхронизированным с вращением Земли. Точное значение в секундах СИ можно лишь оценить, например по данным работы [Измерение вращения Земли: с 720 г. до н. э. по 2015 г. н. э.; 2016, Proceedings A Королевского общества, 472, Stephenson и др.](https://doi.org/10.1098/rspa.2016.0404) («Измерение вращения Земли: с 720 года до н. э. по 2015 год н. э.»). Разумная оценка — `50491112870 ± 90` секунд; разница составляет 10330 секунд.
