```python
import numpy as np
a = np.array([[1,2,3], [4,5,6]])
a
```

```python
np.cumsum(a)
```

```python
np.cumsum(a, dtype=np.float64)  # specifies type of output value(s)
```

**Комментарии к примеру на русском**

Задаёт тип выходных значений

```python
np.cumsum(a,axis=0)      # sum over rows for each of the 3 columns
```

**Комментарии к примеру на русском**

Суммируем по строкам для каждого из трёх столбцов

```python
np.cumsum(a,axis=1)      # sum over columns for each of the 2 rows
```

**Комментарии к примеру на русском**

Суммируем по столбцам для каждой из двух строк

``cumsum(b)[-1]`` may not be equal to ``sum(b)``


```python
b = np.array([1, 2e-9, 3e-9] * 1000000)
b.cumsum()[-1]
```

```python
b.sum()
```
