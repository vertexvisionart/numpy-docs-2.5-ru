Обычное скалярное произведение векторов:


```python
import numpy as np
a = np.array([1,2,3])
b = np.array([0,1,0])
np.inner(a, b)
```

Примеры для многомерных массивов:


```python
a = np.arange(24).reshape((2,3,4))
b = np.arange(4)
c = np.inner(a, b)
c.shape
```

```python
c
```

```python
a = np.arange(2).reshape((1,1,2))
b = np.arange(6).reshape((3,2))
c = np.inner(a, b)
c.shape
```

```python
c
```

Пример, в котором `b` — скаляр:

```python
np.inner(np.eye(2), 7)
```
