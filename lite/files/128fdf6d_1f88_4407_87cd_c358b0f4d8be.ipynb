```python
import numpy as np
```

Создадим массив 4 × 4.


```python
a = np.arange(16).reshape(4, 4)
a
```

Передайте массив, чтобы получить индексы нижних треугольных элементов.

```python
trili = np.tril_indices_from(a)
trili
```

```python
a[trili]
```

Это сокращённая запись вызова tril_indices().


```python
np.tril_indices(a.shape[0])
```

Используем параметр `k`, чтобы получить индексы нижней треугольной части
массива до k-й диагонали включительно.


```python
trili1 = np.tril_indices_from(a, k=1)
a[trili1]
```
