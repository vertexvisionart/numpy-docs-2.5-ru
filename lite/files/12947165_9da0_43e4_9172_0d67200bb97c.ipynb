### DTypeLike

Объединение типов `typing.Union`, описывающее объекты, которые можно преобразовать в `numpy.dtype`. Сюда, в частности, входят:

- объекты `type`;
- символьные коды или имена объектов `type`;
- объекты с атрибутом ``.dtype``.

Добавлено в версии 1.20.

См. также раздел «Задание и создание типов данных»: в нём подробно перечислены все объекты, которые можно преобразовать в типы данных.

Пример:

```python
>>> import numpy as np
>>> import numpy.typing as npt

>>> def as_dtype(d: npt.DTypeLike) -> np.dtype:
...     return np.dtype(d)
```

### NDArray

Псевдоним типа `np.ndarray[tuple[Any, ...], np.dtype[ScalarT]]`, обобщённый по параметру `dtype.type`. Используется для аннотации массивов с заданным dtype и неопределённой формой.

Добавлено в версии 1.21.

Примеры:

```python
>>> import numpy as np
>>> import numpy.typing as npt

>>> print(npt.NDArray)
NDArray

>>> print(npt.NDArray[np.float64])
NDArray[numpy.float64]

>>> NDArrayInt = npt.NDArray[np.int_]
>>> a: NDArrayInt = np.arange(10)

>>> def func(a: npt.ArrayLike) -> npt.NDArray[Any]:
...     return np.array(a)
```
