```python
import numpy as np
import numpy.ma as ma
a = ma.array([[1, 2] ,[3, 4]])
a[0, 1] = ma.masked
a
```

```python
np.resize(a, (3, 3))
```

```python
ma.resize(a, (3, 3))
```

MaskedArray всегда возвращается, независимо от типа ввода.
