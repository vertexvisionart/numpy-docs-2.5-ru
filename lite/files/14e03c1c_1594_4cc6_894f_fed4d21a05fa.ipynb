```python
from numpy.polynomial import legendre as L
c1 = (1,2,3)
c2 = (3,2,1)
L.legdiv(c1,c2) # quotient "intuitive," remainder not
```

**Комментарии к примеру на русском**

Частное «интуитивно понятно», остаток — нет

```python
c2 = (0,1,2,3)
L.legdiv(c2,c1) # neither "intuitive"
```
