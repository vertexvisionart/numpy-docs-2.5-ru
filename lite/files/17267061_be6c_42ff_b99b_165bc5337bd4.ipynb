```python
import numpy as np
np.isneginf(-np.inf)
```

```python
np.isneginf(np.inf)
```

```python
np.isneginf([-np.inf, 0., np.inf])
```

```python
x = np.array([-np.inf, 0., np.inf])
y = np.array([2, 2, 2])
np.isneginf(x, y)
```
