```python
import numpy as np
```

```python
p1d = np.poly1d([1, 2, 3])
p = np.polynomial.Polynomial(p1d.coef[::-1])
```

Помимо атрибута ``coef`` у многочленов из пакета polynomial есть атрибуты ``domain`` и ``window``. Они особенно важны при аппроксимации данных. Многочлены с разными значениями ``domain`` и ``window`` не считаются равными и не могут совместно участвовать в арифметических операциях:


```python
p1 = np.polynomial.Polynomial([1, 2, 3])
p1
```

```python
p2 = np.polynomial.Polynomial([1, 2, 3], domain=[-2, 2])
p1 == p2
```

```python
p1 + p2
```
