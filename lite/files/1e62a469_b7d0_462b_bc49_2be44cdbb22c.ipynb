```python
import numpy as np
a = np.array([[1, 2], [3, 4], [5, 6]])
a
```

```python
np.compress([0, 1], a, axis=0)
```

```python
np.compress([False, True, True], a, axis=0)
```

```python
np.compress([False, True], a, axis=1)
```

При работе с массивом, развёрнутым в одномерный, выбираются отдельные элементы, а не срезы вдоль оси.


```python
np.compress([False, True], a)
```
