Подберём прямую ``y = mx + c`` по точкам с шумом:


```python
import numpy as np
x = np.array([0, 1, 2, 3])
y = np.array([-1, 0.2, 0.9, 2.1])
```

По коэффициентам видно, что наклон прямой примерно равен 1, а пересечение с осью y находится примерно в -1.

Уравнение можно переписать как ``y = Ap``, где ``A = [[x 1]]`` и ``p = [[m], [c]]``. Используем `lstsq` для нахождения `p`:


```python
A = np.vstack([x, np.ones(len(x))]).T
A
```

```python
m, c = np.linalg.lstsq(A, y)[0]
m, c
```

Постройте график данных вместе с подобранной прямой:

```python
import matplotlib.pyplot as plt
_ = plt.plot(x, y, 'o', label='Original data', markersize=10)
_ = plt.plot(x, m*x + c, 'r', label='Fitted line')
_ = plt.legend()
plt.show()
```
