```python
import numpy as np
a = np.arange(9).reshape((3,3))
a = np.ma.array(a)
a[1, 0] = np.ma.masked
a[1, 2] = np.ma.masked
a[2, 1] = np.ma.masked
a
```

```python
np.ma.count_masked(a)
```

При использовании ключевого слова `axis` возвращается массив.

```python
np.ma.count_masked(a, axis=0)
```

```python
np.ma.count_masked(a, axis=1)
```
