Эти примеры демонстрируют низкоуровневый конструктор `ndarray`. Более простые способы создания ndarray см. выше в разделе `See Also`.

Первый режим: `buffer` равен None:

```python
import numpy as np
np.ndarray(shape=(2,2), dtype=np.float64, order='F')
```

Второй режим:

```python
np.ndarray((2,), buffer=np.array([1,2,3]),
           offset=np.int_().itemsize,
           dtype=np.int_) # offset = 1*itemsize, i.e. skip first element
```

**Комментарии к примеру на русском**

offset = 1*itemsize, то есть пропускаем первый элемент
