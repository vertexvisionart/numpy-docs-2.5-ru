Преобразуйте список в массив:

```python
a = [1, 2]
import numpy as np
np.asanyarray(a)
```

Экземпляры подклассов `ndarray` передаются без изменений:

```python
a = np.array([(1., 2), (3., 4)], dtype='f4,i4').view(np.recarray)
np.asanyarray(a) is a
```
