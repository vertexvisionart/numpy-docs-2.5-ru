```python
import numpy as np
from numpy.linalg import inv
a = np.array([[1., 2.], [3., 4.]])
ainv = inv(a)
np.allclose(a @ ainv, np.eye(2))
```

```python
np.allclose(ainv @ a, np.eye(2))
```

Если a — объект matrix, возвращаемое значение также является матрицей:


```python
ainv = inv(np.matrix(a))
ainv
```

Можно одновременно вычислить обратные матрицы для нескольких матриц:

```python
a = np.array([[[1., 2.], [3., 4.]], [[1, 3], [3, 5]]])
inv(a)
```

Если матрица близка к вырожденной, вычисленная обратная матрица может не удовлетворять равенству ``a @ ainv = ainv @ a = eye(a.shape[0])``, даже если исключение `LinAlgError` не возникло:


```python
a = np.array([[2,4,6],[2,0,2],[6,8,14]])
inv(a)  # No errors raised
```

Чтобы обнаружить плохо обусловленную матрицу, вычислите её *число обусловленности* [1] с помощью `numpy.linalg.cond`. Чем оно больше, тем хуже обусловлена матрица. Приближённое правило: если ``cond(a) = 10**k``, можно потерять до ``k`` цифр точности дополнительно к потерям, связанным с численным методом и погрешностями арифметических операций.


```python
from numpy.linalg import cond
cond(a)
```

Плохую обусловленность можно также обнаружить, непосредственно изучив сингулярные значения матрицы. Отношение наибольшего сингулярного значения к наименьшему — это число обусловленности:


```python
from numpy.linalg import svd
sigma = svd(a, compute_uv=False)  # Do not compute singular vectors
sigma.max()/sigma.min()
```
