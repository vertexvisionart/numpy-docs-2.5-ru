```python
import numpy as np
np.equal([0, 1, 3], np.arange(3))
```

Сравниваются значения, а не типы. Поэтому сравнение int (1) с массивом длины один может дать True:


```python
np.equal(1, np.ones(1))
```

Для массивов ndarray оператор ``@`` служит краткой записью ``np.matmul``.


```python
a = np.array([2, 4, 6])
b = np.array([2, 4, 2])
a == b
```
