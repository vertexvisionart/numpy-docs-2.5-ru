Для одномерного массива ``a.tolist()`` почти эквивалентно ``list(a)``, но ``tolist`` преобразует скаляры NumPy в скаляры Python:


```python
import numpy as np
a = np.uint32([1, 2])
a_list = list(a)
a_list
```

```python
type(a_list[0])
```

```python
a_tolist = a.tolist()
a_tolist
```

```python
type(a_tolist[0])
```

Кроме того, для двумерного массива ``tolist`` применяется рекурсивно:


```python
a = np.array([[1, 2], [3, 4]])
list(a)
```

```python
a.tolist()
```

Базовым случаем для этой рекурсии является 0D-массив:

```python
a = np.array(1)
list(a)
```

```python
a.tolist()
```
