```python
import numpy as np
x = np.zeros((3, 4, 5))
np.moveaxis(x, 0, -1).shape
```

```python
np.moveaxis(x, -1, 0).shape
```

Все они достигают одного и того же результата:

```python
np.transpose(x).shape
```

```python
np.swapaxes(x, 0, -1).shape
```

```python
np.moveaxis(x, [0, 1], [-1, -2]).shape
```

```python
np.moveaxis(x, [0, 1, 2], [-1, -2, -3]).shape
```
