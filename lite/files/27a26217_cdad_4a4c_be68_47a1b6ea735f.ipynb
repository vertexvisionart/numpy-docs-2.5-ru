Определить ранг матрицы с помощью сингулярных значений:

```python
s = np.linalg.svdvals([[1, 2, 3],
                       [2, 4, 6],
                       [-1, 1, -1]]); s
```

```python
np.count_nonzero(s > 1e-10)  # Matrix of rank 2
```
