```python
import numpy as np
c = np.array(['aAaAaA', '  aA  ', 'abBABba'])
c
```

```python
np.strings.lstrip(c, 'a')
```

```python
np.strings.lstrip(c, 'A') # leaves c unchanged
```

**Комментарии к примеру на русском**

Оставляет c без изменений

```python
(np.strings.lstrip(c, ' ') == np.strings.lstrip(c, '')).all()
```

```python
(np.strings.lstrip(c, ' ') == np.strings.lstrip(c)).all()
```
