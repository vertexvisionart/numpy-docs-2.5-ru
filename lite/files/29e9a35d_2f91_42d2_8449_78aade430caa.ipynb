```python
import numpy as np
a = np.eye(2*3*4).reshape((2*3, 4, 2, 3, 4))
rng = np.random.default_rng()
b = rng.normal(size=(2*3, 4))
x = np.linalg.tensorsolve(a, b)
x.shape
```

```python
np.allclose(np.tensordot(a, x, axes=3), b)
```
