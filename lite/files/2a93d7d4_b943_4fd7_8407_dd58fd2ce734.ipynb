```python
import numpy as np
x = np.array([1, 2, 3, 5])
N = 3
np.vander(x, N)
```

```python
np.column_stack([x**(N-1-i) for i in range(N)])
```

```python
x = np.array([1, 2, 3, 5])
np.vander(x)
```

```python
np.vander(x, increasing=True)
```

Определитель квадратной матрицы Вандермонда равен произведению разностей значений входного вектора:


```python
np.linalg.det(np.vander(x))
```
