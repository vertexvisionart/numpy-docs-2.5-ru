Сгенерируем значения из распределения и построим гистограмму


```python
from matplotlib.pyplot import hist
values = hist(np.random.rayleigh(3, 100000), bins=200, density=True)
```

Высота волн, как правило, соответствует распределению Рэлея. Если средняя высота волны составляет 1 метр, какая доля волн, вероятно, будет больше 3 метров?

```python
meanvalue = 1
modevalue = np.sqrt(2 / np.pi) * meanvalue
s = np.random.rayleigh(modevalue, 1000000)
```

Процент волн высотой более 3 метров составляет:


```python
100.*sum(s>3)/1000000.
```
