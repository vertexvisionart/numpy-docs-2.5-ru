`default_rng` — рекомендуемый конструктор класса генератора случайных чисел `Generator`. Ниже показано несколько способов создать генератор с помощью `default_rng` и класса `Generator`.

Здесь `default_rng` используется для генерации случайного числа с плавающей точкой:

```python
import numpy as np
rng = np.random.default_rng(12345)
print(rng)
```

```python
rfloat = rng.random()
rfloat
```

```python
type(rfloat)
```

Здесь `default_rng` используется для генерации трёх случайных целых чисел от 0
включительно до 10 не включительно:


```python
import numpy as np
rng = np.random.default_rng(12345)
rints = rng.integers(low=0, high=10, size=3)
rints
```

```python
type(rints[0])
```

Здесь мы задаём начальное состояние генератора для воспроизводимости результатов:


```python
import numpy as np
rng = np.random.default_rng(seed=42)
print(rng)
```

```python
arr1 = rng.random((3, 3))
arr1
```

Если мы выйдем и перезапустим наш интерпретатор Python, мы увидим, что снова генерируем одни и те же случайные числа:

```python
import numpy as np
rng = np.random.default_rng(seed=42)
arr2 = rng.random((3, 3))
arr2
```
