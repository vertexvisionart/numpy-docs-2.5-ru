```python
import numpy as np
np.logical_xor(True, False)
```

```python
np.logical_xor([True, True, False, False], [True, False, True, False])
```

```python
x = np.arange(5)
np.logical_xor(x < 1, x > 3)
```

Простой пример, показывающий поддержку трансляции массивов

```python
np.logical_xor(0, np.eye(2))
```
