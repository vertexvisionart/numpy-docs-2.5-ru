```python
import numpy as np
a = np.arange(4).reshape((2,2))
a
```

```python
np.max(a)           # Maximum of the flattened array
```

**Комментарии к примеру на русском**

Максимум по массиву, развёрнутому в одномерный

```python
np.max(a, axis=0)   # Maxima along the first axis
```

**Комментарии к примеру на русском**

Максимумы вдоль первой оси

```python
np.max(a, axis=1)   # Maxima along the second axis
```

**Комментарии к примеру на русском**

Максимумы вдоль второй оси

```python
np.max(a, where=[False, True], initial=-1, axis=0)
```

```python
b = np.arange(5, dtype=np.float64)
b[2] = np.nan
np.max(b)
```

```python
np.max(b, where=~np.isnan(b), initial=-1)
```

```python
np.nanmax(b)
```

Начальное значение позволяет вычислить максимум пустого среза или задать другое исходное значение:

```python
np.max([[-50], [10]], axis=-1, initial=0)
```

Обратите внимание: начальное значение участвует в определении максимума наравне с элементами массива. Это отличается от аргумента default функции max в Python, который используется только для пустых итерируемых объектов.

```python
np.max([5], initial=6)
```

```python
max([5], default=6)
```
