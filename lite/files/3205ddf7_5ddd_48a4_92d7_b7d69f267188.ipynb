Создадим объект ctypes из ``np.array`` с автоматически определённым целочисленным типом:


```python
inferred_int_array = np.array([1, 2, 3])
c_int_array = np.ctypeslib.as_ctypes(inferred_int_array)
type(c_int_array)
```

```python
c_int_array[:]
```

Создадим объект ctypes из ``np.array`` с явно заданным 8-битным беззнаковым целочисленным типом:


```python
exp_int_array = np.array([1, 2, 3], dtype=np.uint8)
c_int_array = np.ctypeslib.as_ctypes(exp_int_array)
type(c_int_array)
```

```python
c_int_array[:]
```
