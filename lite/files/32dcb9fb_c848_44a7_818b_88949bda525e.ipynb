```python
import numpy as np
a = np.arange(10)
a
```

```python
np.where(a < 5, a, 10*a)
```

Это также может быть использовано в многомерных массивах:

```python
np.where([[True, False], [True, True]],
         [[1, 2], [3, 4]],
         [[9, 8], [7, 6]])
```

Формы x, y и условия согласовываются с помощью трансляции массивов:

```python
x, y = np.ogrid[:3, :4]
np.where(x < y, x, 10 + y)  # both x and 10+y are broadcast
```

**Комментарии к примеру на русском**

И x, и 10+y согласуются по правилам трансляции массивов

```python
a = np.array([[0, 1, 2],
              [0, 2, 4],
              [0, 3, 6]])
np.where(a < 4, a, -1)  # -1 is broadcast
```
