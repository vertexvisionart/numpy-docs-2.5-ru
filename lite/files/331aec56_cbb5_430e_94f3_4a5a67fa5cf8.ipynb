Сгенерируем выборку из распределения:


```python
rng = np.random.default_rng()
n, p, size = 10, .5, 10000
s = rng.binomial(n, p, 10000)
```

Предположим, компания бурит 9 разведочных нефтяных скважин с оценочной вероятностью успеха ``p=0.1`` для каждой. Все девять оказались неудачными. Какова вероятность такого события?

По ``size = 20,000`` испытаниям средняя оценка этой вероятности составляет:


```python
n, p, size = 9, 0.1, 20000
np.sum(rng.binomial(n=n, p=p, size=size) == 0)/size
```

Следующий код визуализирует выборку при ``n=100``, ``p=0.4`` и соответствующую функцию плотности вероятности:


```python
import matplotlib.pyplot as plt
from scipy.stats import binom
n, p, size = 100, 0.4, 10000
sample = rng.binomial(n, p, size=size)
count, bins, _ = plt.hist(sample, 30, density=True)
x = np.arange(n)
y = binom.pmf(x, n, p)
plt.plot(x, y, linewidth=2, color='r')
```
