```python
import numpy as np
a = [4, 3, 5, 7, 6, 8]
indices = [0, 1, 4]
np.take(a, indices)
```

В этом примере, если `a` — ndarray, можно использовать расширенную индексацию.


```python
a = np.array(a)
a[indices]
```

Если `indices` не является одномерным, на выходе также есть эти измерения.

```python
np.take(a, [[0, 1], [2, 3]])
```
