```python
import numpy as np
np.array([1, 2, 3])
```

Приведение к более широкому типу:


```python
np.array([1, 2, 3.0])
```

Более одного измерения:


```python
np.array([[1, 2], [3, 4]])
```

Минимальное число измерений — 2:


```python
np.array([1, 2, 3], ndmin=2)
```

Тип задан:


```python
np.array([1, 2, 3], dtype=np.complex128)
```

Тип данных, состоящий из более чем одного элемента:

```python
x = np.array([(1,2),(3,4)],dtype=[('a','<i4'),('b','<i4')])
x['a']
```

Создание массива из подклассов:

```python
np.array(np.asmatrix('1 2; 3 4'))
```

```python
np.array(np.asmatrix('1 2; 3 4'), subok=True)
```

Ограничение максимального числа измерений с помощью ``ndmax``:


```python
a = np.array([[1, 2], [3, 4]], dtype=np.object_, ndmax=2)
a
```

```python
a.shape
```

```python
b = np.array([[1, 2], [3, 4]], dtype=np.object_, ndmax=1)
b
```

```python
b.shape
```
