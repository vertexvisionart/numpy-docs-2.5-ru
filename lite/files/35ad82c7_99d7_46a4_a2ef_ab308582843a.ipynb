```python
import numpy as np
```

```python
a = np.arange(6).reshape(2,3) - 3
for x in np.nditer(a, op_flags=['readonly','copy'],
                op_dtypes=['complex128']):
    print(np.sqrt(x), end=' ')
```

```python
for x in np.nditer(a, flags=['buffered'], op_dtypes=['complex128']):
    print(np.sqrt(x), end=' ')
```
