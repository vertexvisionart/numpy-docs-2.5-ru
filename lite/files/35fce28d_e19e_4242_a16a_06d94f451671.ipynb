Использование типа скаляра массива:

```python
import numpy as np
np.dtype(np.int16)
```

Структурированный тип с одним полем «f1» типа int16:

```python
np.dtype([('f1', np.int16)])
```

Структурированный тип с одним полем «f1», которое само содержит структурированный тип с одним полем:


```python
np.dtype([('f1', [('f1', np.int16)])])
```

Структурированный тип с двумя полями: первое содержит беззнаковое целое число, второе — int32:


```python
np.dtype([('f1', np.uint64), ('f2', np.int32)])
```

Использование строк типа array-protocol:

```python
np.dtype([('a','f8'),('b','S10')])
```

Форматы полей, перечисленные через запятую. Форма — (2,3):

```python
np.dtype("i4, (2,3)f8")
```

Использование кортежей. ``int`` — тип фиксированного размера, 3 — форма поля. ``void`` — тип переменного размера, в данном случае размером 10:


```python
np.dtype([('hello',(np.int64,3)),('world',np.void,10)])
```

Разделим ``int16`` на два ``int8`` с именами x и y. Числа 0 и 1 — байтовые смещения:


```python
np.dtype((np.int16, {'x':(np.int8,0), 'y':(np.int8,1)}))
```

Использование словарей. Два поля с именами «gender» и «age»:

```python
np.dtype({'names':['gender','age'], 'formats':['S1',np.uint8]})
```

Смещения в байтах, здесь 0 и 25:

```python
np.dtype({'surname':('S25',0),'age':(np.uint8,25)})
```
