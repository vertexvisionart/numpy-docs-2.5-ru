```python
import numpy as np
np.r_[np.array([1,2,3]), 0, 0, np.array([4,5,6])]
```

```python
np.r_[-1:1:6j, [0]*3, 5, 6]
```

Целые числа в строковом аргументе задают ось объединения или минимальное число измерений, к которому нужно привести элементы.


```python
a = np.array([[0, 1, 2], [3, 4, 5]])
np.r_['-1', a, a] # concatenate along last axis
```

**Комментарии к примеру на русском**

Объединяем вдоль последней оси

```python
np.r_['0,2', [1,2,3], [4,5,6]] # concatenate along first axis, dim>=2
```

**Комментарии к примеру на русском**

Объединяем вдоль первой оси, dim>=2

```python
np.r_['0,2,0', [1,2,3], [4,5,6]]
```

```python
np.r_['1,2,0', [1,2,3], [4,5,6]]
```

Использование «r» или «c» в качестве первого аргумента строки создает матрицу.

```python
np.r_['r',[1,2,3], [4,5,6]]
```
