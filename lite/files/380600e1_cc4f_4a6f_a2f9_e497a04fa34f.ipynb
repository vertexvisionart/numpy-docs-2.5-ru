```python
import numpy as np
orig_settings = np.seterr(all='ignore')  # seterr to known value
np.int16(32000) * np.int16(3)
```

```python
old_settings = np.seterr(all='warn', over='raise')
np.int16(32000) * np.int16(3)
```

```python
old_settings = np.seterr(all='print')
np.geterr()
```
