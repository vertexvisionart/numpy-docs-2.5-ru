```python
import numpy as np
np.logical_or(True, False)
```

```python
np.logical_or([True, False], [False, False])
```

```python
x = np.arange(5)
np.logical_or(x < 1, x > 3)
```

Для логических массивов ndarray оператор ``|`` служит краткой записью ``np.logical_or``.


```python
a = np.array([True, False])
b = np.array([False, False])
a | b
```
