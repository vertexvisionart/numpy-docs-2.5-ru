```python
import numpy as np
np.fft.ifft([1, -1j, -1, 1j])
```

```python
np.fft.irfft([1, -1j, -1])
```

Обратите внимание: последний элемент входа обычного `ifft` комплексно сопряжён второму, а мнимая часть результата везде нулевая. При вызове `irfft` отрицательные частоты не задаются, и результат полностью вещественный.
