```python
import numpy as np
a = np.array([[1, 2], [3, np.nan]])
np.nanmax(a)
```

```python
np.nanmax(a, axis=0)
```

```python
np.nanmax(a, axis=1)
```

При наличии положительной и отрицательной бесконечности:


```python
np.nanmax([1, 2, np.nan, -np.inf])
```

```python
np.nanmax([1, 2, np.nan, np.inf])
```
