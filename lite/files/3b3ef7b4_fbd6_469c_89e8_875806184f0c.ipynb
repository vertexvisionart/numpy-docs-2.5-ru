Преобразуйте список в массив:

```python
a = [1, 2]
import numpy as np
np.asarray(a)
```

Существующие массивы не копируются:

```python
a = np.array([1, 2])
np.asarray(a) is a
```

Если `dtype` установлен, массив копируется только в том случае, если dtype не соответствует:

```python
a = np.array([1, 2], dtype=np.float32)
np.shares_memory(np.asarray(a, dtype=np.float32), a)
```

```python
np.shares_memory(np.asarray(a, dtype=np.float64), a)
```

В отличие от `asanyarray`, подклассы ndarray не сохраняются:


```python
issubclass(np.recarray, np.ndarray)
```

```python
a = np.array([(1., 2), (3., 4)], dtype='f4,i4').view(np.recarray)
np.asarray(a) is a
```

```python
np.asanyarray(a) is a
```
