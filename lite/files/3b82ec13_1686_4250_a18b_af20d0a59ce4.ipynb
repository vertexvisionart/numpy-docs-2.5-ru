Примеры одномерных массивов:


```python
import numpy as np
np.add.accumulate([2, 3, 5])
```

```python
np.multiply.accumulate([2, 3, 5])
```

Примеры для двумерных массивов:


```python
I = np.eye(2)
I
```

Накопление вдоль оси 0 (строк), сверху вниз по столбцам:

```python
np.add.accumulate(I, 0)
```

```python
np.add.accumulate(I) # no axis specified = axis zero
```

**Комментарии к примеру на русском**

Ось не указана: используется ось 0

Накопление вдоль оси 1 (столбцов), по каждой строке:


```python
np.add.accumulate(I, 1)
```
