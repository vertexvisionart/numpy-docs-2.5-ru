Сгенерируем выборку из распределения.

Пример из практики: компания бурит поисково-разведочные нефтяные скважины на неизученной территории. Вероятность успеха для каждой оценивается в 0.1. Какова вероятность получить первый успех после бурения пяти скважин, шести скважин и так далее?
Примечание переводчика: код ниже вычисляет вероятность хотя бы одного успеха к этому моменту, то есть первого успеха не позднее указанного числа скважин, а не ровно одного успеха за всё это время.

```python
rng = np.random.default_rng()
s = rng.negative_binomial(1, 0.1, 100000)
for i in range(1, 11): # doctest: +SKIP
   probability = sum(s<i) / 100000.
   print(i, "wells drilled, probability of one success =", probability)
```
