```python
import numpy as np
```

Сохраните данные на диск и загрузите их снова:

```python
np.save('/tmp/123', np.array([[1, 2, 3], [4, 5, 6]]))
np.load('/tmp/123.npy')
```

Сохраним сжатые данные на диск и загрузим их снова:


```python
a=np.array([[1, 2, 3], [4, 5, 6]])
b=np.array([1, 2])
np.savez('/tmp/123.npz', a=a, b=b)
data = np.load('/tmp/123.npz')
data['a']
```

```python
data['b']
```

```python
data.close()
```

Отобразим сохранённый массив в память и получим его вторую строку напрямую с диска:


```python
X = np.load('/tmp/123.npy', mmap_mode='r')
X[1, :]
```
