```python
import numpy as np
```

Первая проверка assert не вызывает исключения:


```python
np.testing.assert_array_equal([1.0,2.33333,np.nan],
                              [np.exp(0),2.33333, np.nan])
```

Проверка завершается ошибкой из-за неточности вычислений с плавающей точкой:


```python
np.testing.assert_array_equal([1.0,np.pi,np.nan],
                              [1, np.sqrt(np.pi)**2, np.nan])
```

В таких случаях используйте `assert_allclose` или одну из функций nulp, сравнивающих значения по числу представимых чисел с плавающей точкой между ними:


```python
np.testing.assert_allclose([1.0,np.pi,np.nan],
                           [1, np.sqrt(np.pi)**2, np.nan],
                           rtol=1e-10, atol=0)
```

Как указано в примечаниях, `assert_array_equal` особым образом обрабатывает скаляры. Здесь проверяется, что каждое значение в `x` равно 3:

```python
x = np.full((2, 5), fill_value=3)
np.testing.assert_array_equal(x, 3)
```

Используйте `strict`, чтобы при сравнении скаляра с массивом возникало AssertionError:


```python
np.testing.assert_array_equal(x, 3, strict=True)
```

Параметр `strict` также гарантирует соответствие типов данных массива:

```python
x = np.array([2, 2, 2])
y = np.array([2., 2., 2.], dtype=np.float32)
np.testing.assert_array_equal(x, y, strict=True)
```
