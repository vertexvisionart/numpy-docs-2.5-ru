Наиболее распространенным применением этой функции является построение блок-матрицы:

```python
import numpy as np
A = np.eye(2) * 2
B = np.eye(3) * 3
np.block([
    [A,               np.zeros((2, 3))],
    [np.ones((3, 2)), B               ]
])
```

Со списком глубины 1 `block` можно использовать как `hstack` :

```python
np.block([1, 2, 3])              # hstack([1, 2, 3])
```

**Комментарии к примеру на русском**

hstack([1, 2, 3])

```python
a = np.array([1, 2, 3])
b = np.array([4, 5, 6])
np.block([a, b, 10])             # hstack([a, b, 10])
```

**Комментарии к примеру на русском**

hstack([a, b, 10])

```python
A = np.ones((2, 2), int)
B = 2 * A
np.block([A, B])                 # hstack([A, B])
```

**Комментарии к примеру на русском**

hstack([A, B])

Для списка с глубиной вложенности 2 `block` можно использовать вместо `vstack`:


```python
a = np.array([1, 2, 3])
b = np.array([4, 5, 6])
np.block([[a], [b]])             # vstack([a, b])
```

**Комментарии к примеру на русском**

vstack([a, b])

```python
A = np.ones((2, 2), int)
B = 2 * A
np.block([[A], [B]])             # vstack([A, B])
```

**Комментарии к примеру на русском**

vstack([A, B])

Он также может быть использован вместо `atleast_1d` и `atleast_2d`:

```python
a = np.array(0)
b = np.array([1])
np.block([a])                    # atleast_1d(a)
```

**Комментарии к примеру на русском**

atleast_1d(a)

```python
np.block([b])                    # atleast_1d(b)
```

**Комментарии к примеру на русском**

atleast_1d(b)

```python
np.block([[a]])                  # atleast_2d(a)
```

**Комментарии к примеру на русском**

atleast_2d(a)

```python
np.block([[b]])                  # atleast_2d(b)
```

**Комментарии к примеру на русском**

atleast_2d(b)
