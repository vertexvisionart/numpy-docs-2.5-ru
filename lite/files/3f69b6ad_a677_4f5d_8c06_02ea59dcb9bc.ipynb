```python
import numpy as np
a = np.ma.array([4, 3, 5, 7, 6, 8], mask=[0, 0, 1, 0, 1, 0])
indices = [0, 1, 4]
np.ma.take(a, indices)
```

Если `indices` не одномерен, результат также содержит его измерения:


```python
np.ma.take(a, [[0, 1], [2, 3]])
```
