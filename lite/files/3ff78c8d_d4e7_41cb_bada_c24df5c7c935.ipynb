Постройте ndarray :

```python
import numpy as np
dt = np.dtype([('time', [('min', np.int64), ('sec', np.int64)]),
               ('temp', float)])
x = np.zeros((1,), dtype=dt)
x['time']['min'] = 10; x['temp'] = 98.25
x
```

Сохранить необработанные данные на диск:

```python
import tempfile
fname = tempfile.mkstemp()[1]
x.tofile(fname)
```

Считаем необработанные данные с диска:


```python
np.fromfile(fname, dtype=dt)
```

Рекомендуемый способ хранения и загрузки данных:

```python
np.save(fname, x)
np.load(fname + '.npy')
```
