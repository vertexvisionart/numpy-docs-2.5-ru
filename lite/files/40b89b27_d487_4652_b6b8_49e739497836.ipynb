Истинное значение ``exp(1e-10) - 1`` приблизительно равно ``1.00000000005e-10``. Этот пример показывает преимущество expm1 при малых аргументах. Примечание переводчика: в оригинале ошибочно указана точность около 32 значащих цифр. Относительная погрешность приведённого десятичного приближения составляет примерно 1,67·10⁻²¹. Это точность самого приближения, а не вычислений float64: у float64 обычно около 15–16 значащих десятичных цифр.

```python
import numpy as np
np.expm1(1e-10)
```

```python
np.exp(1e-10) - 1
```
