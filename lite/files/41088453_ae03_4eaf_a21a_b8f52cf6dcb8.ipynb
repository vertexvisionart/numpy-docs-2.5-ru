```python
import numpy as np
a = np.ma.arange(10)
np.ma.flatnotmasked_contiguous(a)
```

```python
mask = (a < 3) | (a > 8) | (a == 5)
a[mask] = np.ma.masked
np.array(a[~a.mask])
```

```python
np.ma.flatnotmasked_contiguous(a)
```

```python
a[:] = np.ma.masked
np.ma.flatnotmasked_contiguous(a)
```
