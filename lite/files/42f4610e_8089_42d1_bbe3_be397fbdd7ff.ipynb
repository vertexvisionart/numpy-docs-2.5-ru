```python
import numpy.ma as ma
x = [[-1., 2.5], [4., -2.], [3., 0.]]
mask = [[0, 0], [1, 0], [1, 0]]
masked_x = ma.masked_array(x, mask)
masked_x
```

```python
ma.max(masked_x)
```

```python
ma.max(masked_x, axis=0)
```

```python
ma.max(masked_x, axis=1, keepdims=True)
```

```python
mask = [[1, 1], [1, 1], [1, 1]]
masked_x = ma.masked_array(x, mask)
ma.max(masked_x, axis=1)
```
