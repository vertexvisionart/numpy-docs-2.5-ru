```python
import numpy as np
a = np.array([[1,4],[3,1]])
np.sort(a)                # sort along the last axis
```

**Комментарии к примеру на русском**

Сортируем вдоль последней оси

```python
np.sort(a, axis=None)     # sort the flattened array
```

**Комментарии к примеру на русском**

Сортируем массив, развёрнутый в одномерный

```python
np.sort(a, axis=0)        # sort along the first axis
```

**Комментарии к примеру на русском**

Сортируем вдоль первой оси

Используйте ключевое слово `order`, чтобы указать поле для использования при сортировке структурированного массива:

```python
dtype = [('name', 'S10'), ('height', float), ('age', int)]
values = [('Arthur', 1.8, 41), ('Lancelot', 1.9, 38),
          ('Galahad', 1.7, 38)]
a = np.array(values, dtype=dtype)       # create a structured array
np.sort(a, order='height')                        # doctest: +SKIP
```

**Комментарии к примеру на русском**

Создаём структурированный массив

doctest: +SKIP

Сортировать по возрасту, затем по росту, если возрасты равны:

```python
np.sort(a, order=['age', 'height'])               # doctest: +SKIP
```
