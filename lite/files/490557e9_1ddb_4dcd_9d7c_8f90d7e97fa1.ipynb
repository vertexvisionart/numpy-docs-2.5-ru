```python
import numpy as np
```

Определим функцию знака: -1 при ``x < 0`` и +1 при ``x >= 0``.


```python
x = np.linspace(-2.5, 2.5, 6)
np.piecewise(x, [x < 0, x >= 0], [-1, 1])
```

Определим абсолютное значение: ``-x`` при ``x <0`` и ``x`` при ``x >= 0``.


```python
np.piecewise(x, [x < 0, x >= 0], [lambda x: -x, lambda x: x])
```

Применим ту же функцию к скалярному значению.


```python
y = -2
np.piecewise(y, [y < 0, y >= 0], [lambda x: -x, lambda x: x])
```
