В примере из Википедии это распределение можно использовать для разрезания нитей исходной длины 1.0 на K частей разной длины. Для каждой части задаётся средняя длина, но относительные размеры могут случайно меняться.

```python
s = np.random.dirichlet((10, 5, 3), 20).transpose()
```

```python
import matplotlib.pyplot as plt
plt.barh(range(20), s[0])
plt.barh(range(20), s[1], left=s[0], color='g')
plt.barh(range(20), s[2], left=s[0]+s[1], color='r')
plt.title("Lengths of Strings")
```
