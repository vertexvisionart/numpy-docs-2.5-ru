```python
import numpy as np
array_1d = np.arange(10)
np.cumsum(array_1d, axis=1)
```

Сохранились отрицательные оси:

```python
np.cumsum(array_1d, axis=-2)
```

Конструктор класса обычно принимает номер оси и число измерений массива:


```python
print(np.exceptions.AxisError(2, 1, msg_prefix='error'))
```

Также можно передать собственное сообщение об исключении:


```python
print(np.exceptions.AxisError('Custom error message'))
```
