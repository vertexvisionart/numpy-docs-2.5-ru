```python
import numpy as np
from numpy.polynomial import polynomial as P
x = np.linspace(-1,1,51)  # x "data": [-1, -0.96, ..., 0.96, 1]
rng = np.random.default_rng()
err = rng.normal(size=len(x))
y = x**3 - x + err  # x^3 - x + Gaussian noise
c, stats = P.polyfit(x,y,3,full=True)
c # c[0], c[1] approx. -1, c[2] should be approx. 0, c[3] approx. 1
```

**Комментарии к примеру на русском**

Данные x: [-1, -0.96, ..., 0.96, 1]

x^3 - x + гауссовский шум

c[0] и c[1] примерно равны -1, c[2] должен быть близок к 0, c[3] — к 1

```python
stats # note the large SSR, explaining the rather poor results
```

**Комментарии к примеру на русском**

Большая сумма квадратов остатков (SSR) объясняет низкое качество результата

То же самое без добавленного шума


```python
y = x**3 - x
c, stats = P.polyfit(x,y,3,full=True)
c # c[0], c[1] ~= -1, c[2] should be "very close to 0", c[3] ~= 1
```

**Комментарии к примеру на русском**

c[0], c[1] ~= -1, c[2] должен быть очень близок к 0, c[3] ~= 1

```python
stats # note the minuscule SSR
```
