```python
import numpy as np
a = np.array([1,2,3])
np.cumprod(a) # intermediate results 1, 1*2
              # total product 1*2*3 = 6
```

**Комментарии к примеру на русском**

Промежуточные результаты: 1, 1*2

Итоговое произведение: 1*2*3 = 6

```python
a = np.array([[1, 2, 3], [4, 5, 6]])
np.cumprod(a, dtype=np.float64)  # specify type of output
```

**Комментарии к примеру на русском**

Задаём тип результата

Накопленное произведение для каждого столбца `a` (то есть вдоль строк):


```python
np.cumprod(a, axis=0)
```

Накопленное произведение для каждой строки `a` (то есть вдоль столбцов):


```python
np.cumprod(a,axis=1)
```
