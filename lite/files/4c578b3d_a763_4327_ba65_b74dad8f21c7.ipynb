```python
import numpy as np
```

Ожидается, что арккосинус 1 равен 0, а арккосинус -1 — pi:

```python
np.arccos([1, -1])
```

Построим график arccos:


```python
import matplotlib.pyplot as plt
x = np.linspace(-1, 1, num=100)
plt.plot(x, np.arccos(x))
plt.axis('tight')
plt.show()
```
