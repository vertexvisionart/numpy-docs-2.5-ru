```python
import numpy as np
np.union1d([-1, 0, 1], [-2, 0, 2])
```

Чтобы найти объединение более двух массивов, используйте functools.reduce:

```python
from functools import reduce
reduce(np.union1d, ([1, 3, 4, 3], [3, 1, 2, 1], [6, 3, 4, 2]))
```
