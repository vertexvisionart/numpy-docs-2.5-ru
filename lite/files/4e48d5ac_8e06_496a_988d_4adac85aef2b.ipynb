```python
import numpy as np
a = np.array([1, 2])
a.fill(0)
a
```

```python
a = np.empty(2)
a.fill(1)
a
```

Метод fill ожидает скалярное значение и всегда ведёт себя как присваивание одному элементу массива. Ниже приведён редкий пример, в котором это различие важно:

```python
a = np.array([None, None], dtype=np.object_)
a[0] = np.array(3)
a
```

```python
a.fill(np.array(3))
a
```

Тогда как другие формы присваивания распаковывают присваиваемый массив:

```python
a[...] = np.array(3)
a
```
