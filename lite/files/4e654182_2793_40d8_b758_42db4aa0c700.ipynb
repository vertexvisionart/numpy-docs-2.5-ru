```python
a = np.array([1, 2, 3])
np.cumulative_prod(a)  # intermediate results 1, 1*2
                       # total product 1*2*3 = 6
```

**Комментарии к примеру на русском**

Промежуточные результаты: 1, 1*2

Итоговое произведение: 1*2*3 = 6

```python
a = np.array([1, 2, 3, 4, 5, 6])
np.cumulative_prod(a, dtype=np.float64)  # specify type of output
```

**Комментарии к примеру на русском**

Задаём тип результата

Накопленное произведение для каждого столбца ``b`` (то есть вдоль строк):


```python
b = np.array([[1, 2, 3], [4, 5, 6]])
np.cumulative_prod(b, axis=0)
```

Накопленное произведение для каждой строки ``b`` (то есть вдоль столбцов):


```python
np.cumulative_prod(b, axis=1)
```
