```python
import numpy as np
a = np.array([1+1j, 1+0j, 4.5, 3, 2, 2j], dtype=np.complex128)
np.isreal(a)
```

Функция не работает со строковыми массивами.

```python
a = np.array([2j, "a"], dtype=np.str_)
np.isreal(a)  # returns the result of `"" == 0` currently.
```

**Комментарии к примеру на русском**

Сейчас возвращает результат `"" == 0`.

Возвращает True для всех элементов, у которых нет атрибута ``.imag`` либо он равен нулю:


```python
a = np.array([1, "2", 3+4j], dtype=np.object_)
np.isreal(a)
```
