```python
import numpy as np
np.binary_repr(3)
```

```python
np.binary_repr(-3)
```

```python
np.binary_repr(3, width=4)
```

Для отрицательного входного числа при заданном width возвращается представление в дополнительном коде:


```python
np.binary_repr(-3, width=3)
```

```python
np.binary_repr(-3, width=5)
```
