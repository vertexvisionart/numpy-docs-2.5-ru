Это эквивалентно ``reshape(-1, order=order)``.


```python
import numpy as np
x = np.array([[1, 2, 3], [4, 5, 6]])
np.ravel(x)
```

```python
x.reshape(-1)
```

```python
np.ravel(x, order='F')
```

Если ``order`` равен 'A', порядок массива 'C' или 'F' сохраняется:


```python
np.ravel(x.T)
```

```python
np.ravel(x.T, order='A')
```

Если ``order`` равен 'K', сохраняется и порядок, отличный от 'C' и 'F', но оси не разворачиваются:


```python
a = np.arange(3)[::-1]; a
```

```python
a.ravel(order='C')
```

```python
a.ravel(order='K')
```

```python
a = np.arange(12).reshape(2,3,2).swapaxes(1,2); a
```

```python
a.ravel(order='C')
```

```python
a.ravel(order='K')
```
