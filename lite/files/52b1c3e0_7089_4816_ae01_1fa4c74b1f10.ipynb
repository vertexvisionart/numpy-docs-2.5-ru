Пример из практики: в компании работают 10 000 сотрудников службы поддержки,
а среднее время между звонками клиентов составляет 4 минуты.


```python
n = 10000
time_between_calls = np.random.default_rng().exponential(scale=4, size=n)
```

Какова вероятность того, что клиент позвонит в промежутке между 4-й и 5-й минутами от текущего момента?


```python
x = ((time_between_calls < 5).sum())/n
y = ((time_between_calls < 4).sum())/n
x-y
```
