```python
import numpy as np
a = np.zeros((3, 3), int)
np.fill_diagonal(a, 5)
a
```

Эта же функция может работать на 4D-массиве:

```python
a = np.zeros((3, 3, 3, 3), int)
np.fill_diagonal(a, 4)
```

Для ясности мы показываем только несколько блоков:

```python
a[0, 0]
```

```python
a[1, 1]
```

```python
a[2, 2]
```

Параметр wrap влияет только на матрицы, в которых строк больше, чем столбцов:

```python
# tall matrices no wrap
a = np.zeros((5, 3), int)
np.fill_diagonal(a, 4)
a
```

**Комментарии к примеру на русском**

Высокие матрицы без циклического переноса

```python
# tall matrices wrap
a = np.zeros((5, 3), int)
np.fill_diagonal(a, 4, wrap=True)
a
```

**Комментарии к примеру на русском**

Высокие матрицы с циклическим переносом

```python
# wide matrices
a = np.zeros((3, 5), int)
np.fill_diagonal(a, 4, wrap=True)
a
```

**Комментарии к примеру на русском**

Широкие матрицы

Чтобы заполнить побочную диагональ, можно развернуть порядок элементов
с помощью `numpy.flipud` или `numpy.fliplr`.


```python
a = np.zeros((3, 3), int);
np.fill_diagonal(np.fliplr(a), [1,2,3])  # Horizontal flip
a
```

**Комментарии к примеру на русском**

Отражение по горизонтали

```python
np.fill_diagonal(np.flipud(a), [1,2,3])  # Vertical flip
a
```

**Комментарии к примеру на русском**

Отражение по вертикали

Обратите внимание, что порядок, в котором диагональ заполнена, варьируется в зависимости от функции переворачивания.
