```python
import numpy as np
from zoneinfo import ZoneInfo
d = np.arange('2002-10-27T04:30', 4*60, 60, dtype='M8[m]')
d
```

Установка часового пояса в UTC показывает ту же информацию, но с суффиксом Z.

```python
np.datetime_as_string(d, timezone='UTC')
```

Обратите внимание: выбранные даты пересекают границу перехода на летнее время или обратно. При передаче объекта ``ZoneInfo`` выводится соответствующее смещение.


```python
np.datetime_as_string(d, timezone=ZoneInfo('US/Eastern'))
```

Переданная единица времени изменяет точность


```python
np.datetime_as_string(d, unit='h')
```

```python
np.datetime_as_string(d, unit='s')
```

Параметр casting позволяет указать, допустимо ли менять точность


```python
np.datetime_as_string(d, unit='h', casting='safe')
```
