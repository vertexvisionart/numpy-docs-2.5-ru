```python
import numpy as np
```

Преобразуем список в массив. Если все элементы конечны, ``asarray_chkfinite`` эквивалентна ``asarray``.


```python
a = [1, 2]
np.asarray_chkfinite(a, dtype=np.float64)
```

Вызывает ValueError, если array_like содержит NaN или бесконечности.

```python
a = [1, 2, np.inf]
try:
    np.asarray_chkfinite(a)
except ValueError:
    print('ValueError')
```
