````python
import numpy as np
rng = np.random.default_rng()
a = rng.normal(size=(9, 6))
Q, R = np.linalg.qr(a)
np.allclose(a, np.dot(Q, R))  # a does equal QR
````

**Комментарии к примеру на русском**

a действительно равно QR

````python
R2 = np.linalg.qr(a, mode='r')
np.allclose(R, R2)  # mode='r' returns the same R as mode='full'
````

**Комментарии к примеру на русском**

mode='r' возвращает тот же R, что и mode='full'

````python
a = np.random.normal(size=(3, 2, 2)) # Stack of 2 x 2 matrices as input
Q, R = np.linalg.qr(a)
Q.shape
````

**Комментарии к примеру на русском**

На входе — набор матриц 2 x 2

````python
R.shape
````

````python
np.allclose(a, np.matmul(Q, R))
````

Пример распространённого применения `qr` — решение задач методом наименьших квадратов.

Какие `m` и `y0` в уравнении ``y = y0 + mx`` дают наилучшее приближение методом наименьших квадратов для точек {(0,1), (1,0), (1,2), (2,1)}? Построив график, можно увидеть, что должно получиться y0 = 0, m = 1. Ответ находится решением переопределённого матричного уравнения ``Ax = b``, где
```

A = array([[0, 1], [1, 1], [1, 1], [2, 1]])
x = array([[y0], [m]])
b = array([[1], [0], [2], [1]])

```
Если A = QR и Q ортонормальна (такое разложение всегда можно получить методом Грама — Шмидта), то ``x = inv(R) * (Q.T) * b``. На практике в NumPy мы, однако, просто используем `lstsq`.

Примечание переводчика: в исходном примере перепутаны m и y0. Для указанных точек правильные значения — m = 0, y0 = 1; в векторе x коэффициенты также должны идти в порядке [m, y0].


````python
A = np.array([[0, 1], [1, 1], [1, 1], [2, 1]])
A
````

````python
b = np.array([1, 2, 2, 3])
Q, R = np.linalg.qr(A)
p = np.dot(Q.T, b)
np.dot(np.linalg.inv(R), p)
````
