Ручное сложение двух векторов с использованием трансляции массивов:

```python
import numpy as np
x = np.array([[1], [2], [3]])
y = np.array([4, 5, 6])
b = np.broadcast(x, y)
```

```python
out = np.empty(b.shape)
out.flat = [u+v for (u,v) in b]
out
```

Сравним со встроенной трансляцией массивов:


```python
x + y
```
