Сгенерируем 10 000 значений из геометрического распределения с вероятностью успеха в отдельном испытании ``p = 0.35``:


```python
p, size = 0.35, 10000
rng = np.random.default_rng()
sample = rng.geometric(p=p, size=size)
```

Какая доля испытаний завершилась успехом с первой попытки?

```python
(sample == 1).sum()/size
```

Геометрическое распределение с ``p=0.35`` выглядит следующим образом:


```python
import matplotlib.pyplot as plt
count, bins, _ = plt.hist(sample, bins=30, density=True)
plt.plot(bins, (1-p)**(bins-1)*p)
plt.xlim([0, 25])
plt.show()
```
