Создадим объект многочлена ``1 + 7*x + 4*x**2``:


```python
np.polynomial.set_default_printstyle("unicode")
poly = np.polynomial.Polynomial([1, 7, 4])
print(poly)
```

```python
poly.degree()
```

Этот метод не проверяет, какие коэффициенты ненулевые. Чтобы удалить нули в конце, нужно обрезать многочлен:


```python
poly = np.polynomial.Polynomial([1, 7, 0])
print(poly)
```

```python
poly.degree()
```

```python
poly.trim().degree()
```
