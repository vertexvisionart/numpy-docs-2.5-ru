```python
import numpy as np
```

Создаём массив 4 на 4.


```python
a = np.arange(16).reshape(4, 4)
a
```

Получите индексы диагональных элементов.

```python
di = np.diag_indices_from(a)
di
```

```python
a[di]
```

Это лишь синтаксический сахар для diag_indices.


```python
np.diag_indices(a.shape[0])
```
