```python
import numpy as np
```

Это индексы, которые позволят вам получить доступ к верхней треугольной части любого массива 3x3:

```python
iu = np.mask_indices(3, np.triu)
```

Например, если `a` представляет собой массив 3x3:

```python
a = np.arange(9).reshape(3, 3)
a
```

```python
a[iu]
```

В функцию маскирования можно также передать смещение. Так получаем
индексы начиная с первой диагонали справа от главной:


```python
iu1 = np.mask_indices(3, np.triu, 1)
```

С помощью которых мы сейчас извлекаем только три элемента:

```python
a[iu1]
```
