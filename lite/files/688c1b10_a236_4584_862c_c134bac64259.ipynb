Ожидается, что арктангенс 0 равен 0, а арктангенс 1 — pi/4:

```python
import numpy as np
np.arctan([0, 1])
```

```python
np.pi/4
```

График arctan:

```python
import matplotlib.pyplot as plt
x = np.linspace(-10, 10)
plt.plot(x, np.arctan(x))
plt.axis('tight')
plt.show()
```
