```python
import numpy as np
x = np.diag([1.+1.j]*2)
x[1, 1] = 2 + 4.j
x
```

```python
x.getfield(np.float64)
```

Выбрав смещение 8 байт, можно получить представление
мнимой части массива:


```python
x.getfield(np.float64, offset=8)
```
