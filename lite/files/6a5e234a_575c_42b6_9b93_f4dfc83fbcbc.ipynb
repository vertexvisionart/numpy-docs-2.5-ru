````python
import numpy as np
np.histogram([1, 2, 1], bins=[0, 1, 2, 3])
````

````python
np.histogram(np.arange(4), bins=np.arange(5), density=True)
````

````python
np.histogram([[1, 2, 1], [1, 0, 1]], bins=[0,1,2,3])
````

````python
a = np.arange(5)
hist, bin_edges = np.histogram(a, density=True)
hist
````

````python
hist.sum()
````

````python
np.sum(hist * np.diff(bin_edges))
````

Пример автоматического выбора интервалов гистограммы для случайных данных с двумя пиками и 2000 точками.

```python
import matplotlib.pyplot as plt
import numpy as np

rng = np.random.RandomState(10)  # deterministic random data
a = np.hstack((rng.normal(size=1000),
               rng.normal(loc=5, scale=2, size=1000)))
plt.hist(a, bins='auto')  # arguments are passed to np.histogram
plt.title("Histogram with 'auto' bins")
plt.show()
```

Пояснения к примеру: «deterministic random data» — воспроизводимые случайные данные; «arguments are passed to np.histogram» — аргументы передаются в np.histogram. Заголовок графика «Histogram with 'auto' bins» означает «Гистограмма с автоматическим выбором интервалов».
