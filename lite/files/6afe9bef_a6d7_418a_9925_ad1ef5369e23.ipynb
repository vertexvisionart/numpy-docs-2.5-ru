```python
import numpy as np
x = np.array([1, 2, 3, 4])
np.shares_memory(x, np.array([5, 6, 7]))
```

```python
np.shares_memory(x[::2], x)
```

```python
np.shares_memory(x[::2], x[1::2])
```

Проверка того, используют ли два массива общую память, — NP-полная задача. Время выполнения может расти экспоненциально с числом измерений. Поэтому `max_work` обычно следует ограничивать конечным числом: можно построить примеры, проверка которых занимает чрезвычайно много времени:


```python
from numpy.lib.stride_tricks import as_strided
x = np.zeros([192163377], dtype=np.int8)
x1 = as_strided(
    x, strides=(36674, 61119, 85569), shape=(1049, 1049, 1049))
x2 = as_strided(
    x[64023025:], strides=(12223, 12224, 1), shape=(1049, 1049, 1))
np.shares_memory(x1, x2, max_work=1000)
```

В этом случае вызов ``np.shares_memory(x1, x2)`` без заданного `max_work` занимает около минуты. Бывают задачи, требующие значительно больше времени.
