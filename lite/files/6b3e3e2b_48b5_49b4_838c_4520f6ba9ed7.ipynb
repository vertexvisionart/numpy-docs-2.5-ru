```python
import numpy as np
a = np.array([(1, 2.0), (1, 2.0)],
    dtype=[('x', np.int64), ('y', np.float64)])
a
```

Здесь каждый элемент состоит из двух полей: x (int) и y (float).
Такой массив называется структурированным. Поля аналогичны
столбцам электронной таблицы. Обращаться к ним можно так же,
как к элементам словаря:


```python
a['x']
```

```python
a['y']
```

Массивы записей позволяют обращаться к полям как к свойствам:


```python
ar = np.rec.array(a)
ar.x
```

```python
ar.y
```
