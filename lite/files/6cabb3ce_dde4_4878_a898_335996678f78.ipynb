Этот API требует передачи объектов dtype; для удобства определим их:


```python
import numpy as np
int32 = np.dtype(np.int32)
float32 = np.dtype(np.float32)
```

При обычном вызове универсальной функции тип выходных данных не задают. У `numpy.add` два входа и один выход, поэтому оставим тип выхода равным ``None`` (не задан):


```python
np.add.resolve_dtypes((int32, float32, None))
```

Найденный цикл использует float64 для всех операндов, включая выход; первый вход будет приведён к этому типу.

``resolve_dtypes`` поддерживает «слабую» обработку скаляров Python при передаче ``int``, ``float`` или ``complex``:


```python
np.add.resolve_dtypes((float32, float, None))
```

Здесь тип Python ``float`` ведёт себя при вызове ufunc подобно значению Python ``0.0``. Подробности см. в NEP 50.
