Предположим, что компания имеет 10000 агентов поддержки клиентов, и время между вызовами клиентов экспоненциально распределено, а среднее время между вызовами клиентов составляет 4 минуты.

```python
scale, size = 4, 10000
rng = np.random.default_rng()
time_between_calls = rng.exponential(scale=scale, size=size)
```

Какова вероятность того, что клиент позвонит в промежутке между 4-й и 5-й минутами от текущего момента?


```python
x = ((time_between_calls < 5).sum())/size
y = ((time_between_calls < 4).sum())/size
x - y
```

Соответствующее распределение можно визуализировать следующим образом:

```python
import matplotlib.pyplot as plt
scale, size = 4, 10000
rng = np.random.default_rng()
sample = rng.exponential(scale=scale, size=size)
count, bins, _ = plt.hist(sample, 30, density=True)
plt.plot(bins, scale**(-1)*np.exp(-scale**-1*bins), linewidth=2, color='r')
plt.show()
```
