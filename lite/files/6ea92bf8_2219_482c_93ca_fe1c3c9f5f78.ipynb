```python
import numpy as np
```

Двумерная псевдоматрица Вандермонда степени ``[1, 2]`` для точек ``x = [-1, 2]`` и ``y = [1, 3]`` имеет вид:


```python
from numpy.polynomial import polynomial as P
x = np.array([-1, 2])
y = np.array([1, 3])
m, n = 1, 2
deg = np.array([m, n])
V = P.polyvander2d(x=x, y=y, deg=deg)
V
```

Можно проверить столбцы для любых ``0 <= i <= m`` и ``0 <= j <= n``:


```python
i, j = 0, 1
V[:, (deg[1]+1)*i + j] == x**i * y**j
```

Одномерная матрица Вандермонда для точек ``x`` и степени ``m`` — частный случай двумерной псевдоматрицы Вандермонда, у которой все точки ``y`` равны нулю, а степень равна ``[m, 0]``.


```python
P.polyvander2d(x=x, y=0*x, deg=(m, 0)) == P.polyvander(x=x, deg=m)
```
