```python
import numpy as np
```

```python
a = np.arange(3)
b = np.arange(8).reshape(2,4)
it = np.nditer([a, b, None], flags=['external_loop'],
            op_axes=[[0, -1, -1], [-1, 0, 1], None])
with it:
    for x, y, z in it:
        z[...] = x*y
    result = it.operands[2]  # same as z

result
```
