```python
import numpy as np
np.ma.masked_all((3, 3))
```

Параметр `dtype` определяет базовый тип данных.

```python
a = np.ma.masked_all((3, 3))
a.dtype
```

```python
a = np.ma.masked_all((3, 3), dtype=np.int32)
a.dtype
```
