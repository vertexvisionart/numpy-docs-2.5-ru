```python
import numpy as np
np.fromfunction(lambda i, j: i, (2, 2), dtype=np.float64)
```

```python
np.fromfunction(lambda i, j: j, (2, 2), dtype=np.float64)
```

```python
np.fromfunction(lambda i, j: i == j, (3, 3), dtype=np.int_)
```

```python
np.fromfunction(lambda i, j: i + j, (3, 3), dtype=np.int_)
```
