```python
import numpy as np
```

Вычислим два набора индексов для обращения к массивам 4×4: один для нижней треугольной части, начиная с главной диагонали, а другой — начиная с диагонали на две позиции правее:

```python
il1 = np.tril_indices(4)
il1
```

Индексы строк (первый массив) не убывают, а соответствующие
индексы столбцов (второй массив) строго возрастают в пределах каждой строки.
Вот как использовать их на примере массива:


```python
a = np.arange(16).reshape(4, 4)
a
```

Оба для индексации:

```python
a[il1]
```

И для присвоения значений:

```python
a[il1] = -1
a
```

Они охватывают почти весь массив (две диагонали справа от основной):

```python
il2 = np.tril_indices(4, 2)
a[il2] = -10
a
```
