Сгенерируем выборку из распределения:


```python
mu, sigma = 3., 1. # mean and standard deviation
s = np.random.lognormal(mu, sigma, 1000)
```

**Комментарии к примеру на русском**

Среднее и стандартное отклонение

Построим гистограмму выборки вместе
с функцией плотности вероятности:


```python
import matplotlib.pyplot as plt
count, bins, ignored = plt.hist(s, 100, density=True, align='mid')
```

```python
x = np.linspace(min(bins), max(bins), 10000)
pdf = (np.exp(-(np.log(x) - mu)**2 / (2 * sigma**2))
       / (x * sigma * np.sqrt(2 * np.pi)))
```

```python
plt.plot(x, pdf, linewidth=2, color='r')
plt.axis('tight')
plt.show()
```

Покажем, что произведения случайных величин из равномерного распределения хорошо описываются плотностью логнормального распределения.

```python
# Generate a thousand samples: each is the product of 100 random
# values, drawn from a normal distribution.
b = []
for i in range(1000):
   a = 10. + np.random.standard_normal(100)
   b.append(np.prod(a))
```

**Комментарии к примеру на русском**

Создаём тысячу выборочных значений: каждое — произведение 100 случайных

значений из нормального распределения.

```python
b = np.array(b) / np.min(b) # scale values to be positive
count, bins, ignored = plt.hist(b, 100, density=True, align='mid')
sigma = np.std(np.log(b))
mu = np.mean(np.log(b))
```

**Комментарии к примеру на русском**

Масштабируем значения так, чтобы они были положительными

```python
x = np.linspace(min(bins), max(bins), 10000)
pdf = (np.exp(-(np.log(x) - mu)**2 / (2 * sigma**2))
       / (x * sigma * np.sqrt(2 * np.pi)))
```

```python
plt.plot(x, pdf, color='r', linewidth=2)
plt.show()
```
