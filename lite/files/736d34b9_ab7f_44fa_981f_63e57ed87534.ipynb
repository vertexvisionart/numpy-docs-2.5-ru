Одномерный массив:


```python
import numpy as np
x = np.array([3, 1, 2])
np.argsort(x)
```

Двумерный массив:

```python
x = np.array([[0, 3], [2, 2]])
x
```

```python
ind = np.argsort(x, axis=0)  # sorts along first axis (down)
ind
```

**Комментарии к примеру на русском**

Сортирует вдоль первой оси, сверху вниз

```python
np.take_along_axis(x, ind, axis=0)  # same as np.sort(x, axis=0)
```

**Комментарии к примеру на русском**

То же, что np.sort(x, axis=0)

```python
ind = np.argsort(x, axis=1)  # sorts along last axis (across)
ind
```

**Комментарии к примеру на русском**

Сортирует вдоль последней оси, по строкам

```python
np.take_along_axis(x, ind, axis=1)  # same as np.sort(x, axis=1)
```

**Комментарии к примеру на русском**

То же, что np.sort(x, axis=1)

Индексы отсортированных элементов N-мерного массива:


```python
ind = np.unravel_index(np.argsort(x, axis=None), x.shape)
ind
```

```python
x[ind]  # same as np.sort(x, axis=None)
```

**Комментарии к примеру на русском**

То же, что np.sort(x, axis=None)

Сортировка с ключами:

```python
x = np.array([(1, 0), (0, 1)], dtype=[('x', '<i4'), ('y', '<i4')])
x
```

```python
np.argsort(x, order=('x','y'))
```

```python
np.argsort(x, order=('y','x'))
```
