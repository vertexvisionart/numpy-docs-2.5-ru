```python
import numpy as np
```

```python
f16 = np.float16("0.1")
f32 = np.float32(f16)
f64 = np.float64(f32)
f16 == f32 == f64
```

```python
f16, f32, f64
```

Ни одно из этих чисел с плавающей точкой не хранит точное значение $\frac{1}{10}$. Значение ``f16`` выводится как ``0.1``, поскольку это ближайшее к нему представимое число. Другие типы имеют большую точность и позволяют представить ещё более близкие значения, поэтому выводятся иначе.

И наоборот, скаляры с плавающей точкой разной точности, приближающие одно десятичное значение, могут оказаться неравными при сравнении, хотя выводятся одинаково:


```python
f16 = np.float16("0.1")
f32 = np.float32("0.1")
f64 = np.float64("0.1")
f16 == f32 == f64
```

```python
f16, f32, f64
```
