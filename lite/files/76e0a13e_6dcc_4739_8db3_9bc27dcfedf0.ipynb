```python
import numpy as np
np.multiply.reduce([2,3,5])
```

Пример многомерного массива:

```python
X = np.arange(8).reshape((2,2,2))
X
```

```python
np.add.reduce(X, 0)
```

```python
np.add.reduce(X) # confirm: default axis value is 0
```

**Комментарии к примеру на русском**

Проверяем: значение axis по умолчанию равно 0

```python
np.add.reduce(X, 1)
```

```python
np.add.reduce(X, 2)
```

Именованный аргумент ``initial`` позволяет задать другое начальное значение свёртки, а ``where`` — выбрать учитываемые элементы:


```python
np.add.reduce([10], initial=5)
```

```python
np.add.reduce(np.ones((2, 2, 2)), axis=(0, 2), initial=10)
```

```python
a = np.array([10., np.nan, 10])
np.add.reduce(a, where=~np.isnan(a))
```

Позволяет выполнять свёртки пустых массивов там, где они обычно завершались бы ошибкой, то есть для универсальных функций без нейтрального элемента.


```python
np.minimum.reduce([], initial=np.inf)
```

```python
np.minimum.reduce([[1., 2.], [3., 4.]], initial=10., where=[True, False])
```

```python
np.minimum.reduce([])
```
