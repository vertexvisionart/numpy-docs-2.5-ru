```python
import numpy as np
```

Рассмотрим две переменные, $x_0$ и $x_1$, с идеальной отрицательной корреляцией:


```python
x = np.array([[0, 2], [1, 1], [2, 0]]).T
x
```

Заметьте: $x_0$ растёт, а $x_1$ убывает. Это хорошо видно из ковариационной матрицы:


```python
np.cov(x)
```

Обратите внимание, что элемент $C_{0,1}$, который показывает корреляцию между $x_0$ и $x_1$, является отрицательным. Далее, обратите внимание, как `x` и `y` объединены:

```python
x = [-2.1, -1,  4.3]
y = [3,  1.1,  0.12]
X = np.stack((x, y), axis=0)
np.cov(X)
```

```python
np.cov(x, y)
```
