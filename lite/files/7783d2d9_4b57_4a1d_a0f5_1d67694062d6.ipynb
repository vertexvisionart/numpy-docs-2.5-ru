````python
import numpy as np
a = np.array([7, 1, 7, 7, 1, 5, 7, 2, 3, 2, 6, 2, 3, 0])
p = np.partition(a, 4)
p
````

``p[4]`` равно 2; все элементы ``p[:4]`` не превышают ``p[4]``, а все элементы ``p[5:]`` не меньше ``p[4]``. Разбиение имеет вид:
```

[0, 1, 2, 1], [2], [5, 2, 3, 3, 6, 7, 7, 7, 7]

```
Следующий пример показывает передачу нескольких значений в `kth`.


````python
p2 = np.partition(a, (4, 8))
p2
````

``p2[4]`` равно 2, а ``p2[8]`` равно 5. Все элементы ``p2[:4]`` не превышают ``p2[4]``; элементы ``p2[5:8]`` не меньше ``p2[4]`` и не больше ``p2[8]``; элементы ``p2[9:]`` не меньше ``p2[8]``. Разбиение имеет вид:
```

[0, 1, 2, 1], [2], [3, 3, 2], [5], [6, 7, 7, 7, 7]
```
