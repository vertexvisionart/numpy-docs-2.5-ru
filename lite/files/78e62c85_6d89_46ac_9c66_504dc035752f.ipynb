```python
import numpy.ma as ma
a = ma.arange(6).reshape((2, 3))
a[1, :] = ma.masked
a
```

```python
a.count()
```

Если задан именованный аргумент `axis`, возвращается массив
соответствующего размера.


```python
a.count(axis=0)
```
