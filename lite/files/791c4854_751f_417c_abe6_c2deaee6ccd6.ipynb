```python
import numpy as np
element = 2*np.arange(4).reshape((2, 2))
element
```

```python
test_elements = [1, 2, 4, 8]
mask = np.isin(element, test_elements)
mask
```

```python
element[mask]
```

Индексы сопоставленных значений можно получить с помощью `nonzero`:

```python
np.nonzero(mask)
```

Условие проверки можно также обратить:


```python
mask = np.isin(element, test_elements, invert=True)
mask
```

```python
element[mask]
```

Из-за особенностей обработки множеств в `array` следующий код работает не так, как ожидается:


```python
test_set = {1, 2, 4, 8}
np.isin(element, test_set)
```

Преобразование множества в список даёт ожидаемый результат:


```python
np.isin(element, list(test_set))
```
