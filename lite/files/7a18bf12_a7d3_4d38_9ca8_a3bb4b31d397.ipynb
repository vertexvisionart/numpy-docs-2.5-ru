Ниже приведён типичный пример: `NBitBase` используется для аннотации функции, которая принимает число с плавающей точкой и целое число произвольной разрядности, а возвращает число с плавающей точкой с большей из двух разрядностей (*например*, ``np.float16 + np.int64 -> np.float64``).

```python
>>> from typing import TYPE_CHECKING
>>> import numpy as np
>>> import numpy.typing as npt

>>> def add[S: npt.NBitBase, T: npt.NBitBase](
...     a: np.floating[S], b: np.integer[T]
... ) -> np.floating[S | T]:
...     return a + b

>>> a = np.float16()
>>> b = np.int64()
>>> out = add(a, b)

>>> if TYPE_CHECKING:
...     reveal_locals()
...     # note: Revealed local types are:
...     # note:     a: numpy.floating[numpy.typing._16Bit*]
...     # note:     b: numpy.signedinteger[numpy.typing._64Bit*]
...     # note:     out: numpy.floating[numpy.typing._64Bit*]
```

Пояснение к комментариям: «Revealed local types are» означает «Определены следующие типы локальных переменных». Переменная a имеет тип numpy.floating с разрядностью _16Bit, b — numpy.signedinteger с разрядностью _64Bit, out — numpy.floating с разрядностью _64Bit. Комментарии показывают вывод анализатора типов; TYPE_CHECKING позволяет выполнять этот разбор при статической проверке.
