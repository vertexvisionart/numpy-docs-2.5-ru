Индексы могут использоваться в качестве индекса в массиве.

```python
x = np.arange(20).reshape(5, 4)
row, col = np.indices((2, 3))
x[row, col]
```

В приведённом выше примере проще извлечь нужные элементы непосредственно с помощью ``x[:2, :3]``.

Если sparse имеет истинное значение, сетка возвращается в разреженном представлении.


```python
i, j = np.indices((2, 3), sparse=True)
i.shape
```

```python
j.shape
```

```python
i        # row indices
```
