Уменьшение массива: он разворачивается в одномерный в порядке расположения данных в памяти, его размер меняется, затем задаётся новая форма:


```python
import numpy as np
```

```python
a = np.array([[0, 1], [2, 3]], order='C')
a.resize((2, 1))
a
```

```python
a = np.array([[0, 1], [2, 3]], order='F')
a.resize((2, 1))
a
```

Расширение массива: как и выше, но недостающие записи заполнены нулями:

```python
b = np.array([[0, 1], [2, 3]])
b.resize(2, 3) # new_shape parameter doesn't have to be a tuple
b
```
