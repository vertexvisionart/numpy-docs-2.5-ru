```python
import numpy as np
import ctypes
x = np.array([[0, 1], [2, 3]], dtype=np.int32)
x
```

```python
x.ctypes.data
```

```python
x.ctypes.data_as(ctypes.POINTER(ctypes.c_uint32))
```

```python
x.ctypes.data_as(ctypes.POINTER(ctypes.c_uint32)).contents
```

```python
x.ctypes.data_as(ctypes.POINTER(ctypes.c_uint64)).contents
```

```python
x.ctypes.shape
```

```python
x.ctypes.strides
```
