Обратите внимание: оператор свёртки переворачивает второй массив перед тем, как «скользить» массивами друг относительно друга:

```python
import numpy as np
np.convolve([1, 2, 3], [0, 1, 0.5])
```

Возвращаются только центральные значения свёртки. Есть граничные эффекты, поскольку учитываются нули:


```python
np.convolve([1,2,3],[0,1,0.5], 'same')
```

Два массива имеют одинаковую длину, поэтому есть только одно положение, где они полностью перекрываются:

```python
np.convolve([1,2,3],[0,1,0.5], 'valid')
```
