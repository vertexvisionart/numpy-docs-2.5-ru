```python
rng = np.random.default_rng()
rng.integers(2, size=10)
```

```python
rng.integers(1, size=10)
```

Создаём массив формы 2 x 4 из целых чисел от 0 до 4 включительно:


```python
rng.integers(5, size=(2, 4))
```

Создайте массив 1 x 3 с 3 различными верхними границами

```python
rng.integers(1, [3, 5, 10])
```

Создайте массив 1 на 3 с 3 различными нижними границами

```python
rng.integers([1, 5, 7], 10)
```

Генерировать массив 2 на 4 с помощью трансляции массивов с dtype uint8

```python
rng.integers([1, 3, 5, 7], [[10], [20]], dtype=np.uint8)
```
