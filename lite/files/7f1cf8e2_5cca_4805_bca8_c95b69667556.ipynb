Создадим массив с двумя полями, ``x`` и ``y``:


```python
import numpy as np
x = np.array([(1.0, 2), (3.0, 4)], dtype=[('x', '<f8'), ('y', '<i8')])
x
```

```python
x['x']
```

Получим представление массива как массива записей:


```python
x = x.view(np.recarray)
```

```python
x.x
```

```python
x.y
```

Создайте новый, пустой массив записей:

```python
np.recarray((2,),
dtype=[('x', int), ('y', float), ('z', int)]) #doctest: +SKIP
```
