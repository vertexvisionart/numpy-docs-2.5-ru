```python
import numpy as np
np.cos(np.array([0, np.pi/2, np.pi]))
```

```python

# Example of providing the optional output parameter
out1 = np.array([0], dtype=np.float64)
out2 = np.cos([0.1], out1)
out2 is out1
```
