```python
import numpy as np
x = np.array([-1.2, 1.2])
np.absolute(x)
```

```python
np.absolute(1.2 + 1j)
```

Построим график функции на интервале ``[-10, 10]``:


```python
import matplotlib.pyplot as plt
```

```python
x = np.linspace(start=-10, stop=10, num=101)
plt.plot(x, np.absolute(x))
plt.show()
```

Построим график функции на комплексной плоскости:


```python
xx = x + 1j * x[:, np.newaxis]
plt.imshow(np.abs(xx), extent=[-10, 10, -10, 10], cmap='gray')
plt.show()
```
