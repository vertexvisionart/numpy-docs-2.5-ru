```python
import numpy as np
for dt in [np.int32, np.int64, np.float64, np.complex128]:
    np.ma.array([0, 1], dtype=dt).get_fill_value()
```

```python
x = np.ma.array([0, 1.], fill_value=-np.inf)
x.fill_value
```

```python
x.fill_value = np.pi
x.fill_value
```

Восстановление настроек по умолчанию:


```python
x.fill_value = None
x.fill_value
```
