Решим систему уравнений:
``x0 + 2 * x1 = 1`` и
``3 * x0 + 5 * x1 = 2``:


```python
import numpy as np
a = np.array([[1, 2], [3, 5]])
b = np.array([1, 2])
x = np.linalg.solve(a, b)
x
```

Проверим правильность решения:


```python
np.allclose(np.dot(a, x), b)
```
