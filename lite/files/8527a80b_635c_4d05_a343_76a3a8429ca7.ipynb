```python
import numpy as np
```

Для этого массива-примера


```python
a = np.array([[10, 30, 20], [60, 40, 50]])
```

Сортировку можно выполнить напрямую с помощью sort либо с помощью argsort и этой функции.


```python
np.sort(a, axis=1)
```

```python
ai = np.argsort(a, axis=1)
ai
```

```python
np.take_along_axis(a, ai, axis=1)
```

То же работает для max и min, если сохранить измерение размера 1 с помощью ``keepdims``:


```python
np.max(a, axis=1, keepdims=True)
```

```python
ai = np.argmax(a, axis=1, keepdims=True)
ai
```

```python
np.take_along_axis(a, ai, axis=1)
```

Чтобы получить максимум и минимум одновременно, сначала можно объединить
индексы:


```python
ai_min = np.argmin(a, axis=1, keepdims=True)
ai_max = np.argmax(a, axis=1, keepdims=True)
ai = np.concatenate([ai_min, ai_max], axis=1)
ai
```

```python
np.take_along_axis(a, ai, axis=1)
```
