```python
from tempfile import TemporaryFile
a = np.empty(10,dtype='f8,i4,S5')
a[5] = (0.5,10,'abcde')

fd=TemporaryFile()
a = a.view(a.dtype.newbyteorder('<'))
a.tofile(fd)

_ = fd.seek(0)
r=np.rec.fromfile(fd, formats='f8,i4,S5', shape=10,
byteorder='<')
print(r[5])
```
