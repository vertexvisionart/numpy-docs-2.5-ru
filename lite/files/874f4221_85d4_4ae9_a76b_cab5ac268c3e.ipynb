```python
import numpy as np
from numpy.lib.stride_tricks import sliding_window_view
x = np.arange(6)
x.shape
```

```python
v = sliding_window_view(x, 3)
v.shape
```

```python
v
```

Это работает и для большего числа измерений, например:

```python
i, j = np.ogrid[:3, :4]
x = 10*i + j
x.shape
```

```python
x
```

```python
shape = (2,2)
v = sliding_window_view(x, shape)
v.shape
```

```python
v
```

Ось можно указать явно:


```python
v = sliding_window_view(x, 3, 0)
v.shape
```

```python
v
```

Одну ось можно указать несколько раз. Каждое такое указание уменьшает
размер соответствующего исходного измерения:


```python
v = sliding_window_view(x, (2, 3), (1, 1))
v.shape
```

```python
v
```

В сочетании со срезами с шагом (`::step`) это позволяет создавать скользящие представления с пропуском элементов:


```python
x = np.arange(7)
sliding_window_view(x, 5)[:, ::2]
```

или представления с шагом в несколько элементов


```python
x = np.arange(7)
sliding_window_view(x, 3)[::2, :]
```

Распространённое применение `sliding_window_view` — вычисление скользящих статистик. Простейший пример — скользящее среднее ([скользящее среднее](https://en.wikipedia.org/wiki/Moving_average)):


```python
x = np.arange(6)
x.shape
```

```python
v = sliding_window_view(x, 3)
v.shape
```

```python
v
```

```python
moving_average = v.mean(axis=-1)
moving_average
```

Чтобы изменить величину шага скользящего окна, проиндексируйте выходное представление
по нужным измерениям. Для показанного выше массива:


```python
v[::2]
```

Та же техника позволяет перемещать окно в обратном направлении:


```python
v[::-1]
```

Следующие два примера показывают действие ``writeable=True``.

Запись в представление, созданное с параметром по умолчанию ``writeable=False``, вызывает ошибку.


```python
v = sliding_window_view(x, 3)
v[0,1] = 10
```

Запись в представление, созданное с ``writeable=True``, изменяет исходный массив и сразу несколько позиций представления.


```python
x = np.arange(6)  # reset x for the second example
v = sliding_window_view(x, 3, writeable=True)
v[0,1] = 10
x
```
