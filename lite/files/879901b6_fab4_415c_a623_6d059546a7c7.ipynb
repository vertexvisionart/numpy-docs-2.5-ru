```python
import numpy as np
x = np.arange(10)
m = np.ma.masked_array(x, x>5)
assert not m.hardmask
```

Поскольку у `m` мягкая маска, присваивание значения элементу снимает с него маску:


```python
m[8] = 42
m
```

После включения жёсткой маски присваивания больше не меняют её:

```python
hardened = np.ma.harden_mask(m)
assert m.hardmask and hardened is m
m[:] = 23
m
```
