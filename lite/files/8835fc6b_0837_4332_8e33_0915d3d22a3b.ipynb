```python
import numpy as np
```

```python
from numpy.lib import recfunctions as rfn
a = np.ones(4, dtype=[('a', 'i4'), ('b', 'f8'), ('c', 'u1')])
rfn.require_fields(a, [('b', 'f4'), ('c', 'u1')])
```

```python
rfn.require_fields(a, [('b', 'f4'), ('newf', 'u1')])
```
