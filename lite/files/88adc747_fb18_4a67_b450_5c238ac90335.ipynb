Тип данных с полями ``r``, ``g``, ``b``, ``a``, каждое из которых — 8-битное беззнаковое целое число:


```python
import numpy as np
```

```python
dt = np.dtype({'names': ['r','g','b','a'],
               'formats': [np.uint8, np.uint8, np.uint8, np.uint8]})
```

Тип данных с полями ``r`` и ``b`` и заданными дополнительными названиями. Оба поля — 8-битные беззнаковые целые числа; первое имеет байтовое смещение 0 от начала записи, второе — смещение 2. Примечание переводчика: оригинал ошибочно говорит «от начала поля»; offsets задаёт смещения полей от начала содержащей их записи:

```python
dt = np.dtype({'names': ['r','b'], 'formats': ['u1', 'u1'],
               'offsets': [0, 2],
               'titles': ['Red pixel', 'Blue pixel']})
```
