Тип данных с полями ``big`` (32-битное целое число с порядком байтов от старшего к младшему) и ``little`` (32-битное целое число с порядком от младшего к старшему):


```python
import numpy as np
```

```python
dt = np.dtype([('big', '>i4'), ('little', '<i4')])
```

Тип данных с полями ``R``, ``G``, ``B``, ``A``, каждое из которых — 8-битное беззнаковое целое число:


```python
dt = np.dtype([('R','u1'), ('G','u1'), ('B','u1'), ('A','u1')])
```
