Бросаем игральную кость 20 раз:

```python
rng = np.random.default_rng()
rng.multinomial(20, [1/6.]*6, size=1)
```

Единица выпала 4 раза, двойка — один раз и т. д.

Теперь бросим кость 20 раз, а затем ещё 20 раз:

```python
rng.multinomial(20, [1/6.]*6, size=2)
```

В первом запуске единица выпала 3 раза, двойка — 4 раза и так далее. Во втором единица выпала 2 раза, двойка — 4 раза и так далее.

Теперь проведём один эксперимент из 10 бросков и ещё 10 бросков, а другой — из 20 бросков и ещё 20 бросков:

```python
rng.multinomial([[10], [20]], [1/6.]*6, size=(2, 2))
```

Первый массив показывает результаты 10 бросков кости, второй — 20 бросков.

У нечестной кости вероятность выпадения 6 повышена:


```python
rng.multinomial(100, [1/7.]*5 + [2/7.])
```

Смоделируем 10 бросков четырёхгранной и 20 бросков шестигранной игральной кости


```python
rng.multinomial([10, 20],[[1/4]*4 + [0]*2, [1/6]*6])
```

Генерирует категориальные случайные величины для двух категорий: у первой 3 возможных исхода, у второй — 2.

```python
rng.multinomial(1, [[.1, .5, .4 ], [.3, .7, .0]])
```

Затем для получения категорий используется ``argmax(axis=-1)``.


```python
pvals = [[.1, .5, .4 ], [.3, .7, .0]]
rvs = rng.multinomial(1, pvals, size=(4,2))
rvs.argmax(axis=-1)
```

Ту же размерность результата можно получить с помощью трансляции массивов.

```python
rvs = rng.multinomial([[1]] * 4, pvals)
rvs.argmax(axis=-1)
```

Входные вероятности должны быть нормированы. Деталь реализации: значение последнего элемента игнорируется и считается равным оставшейся вероятностной массе, но полагаться на это не следует. Моделировать броски несимметричной монеты, у которой одна сторона вдвое вероятнее другой, нужно так:


```python
rng.multinomial(100, [1.0 / 3, 2.0 / 3])  # RIGHT
```
