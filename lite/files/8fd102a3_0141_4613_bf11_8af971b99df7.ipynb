```python
import numpy as np
```

```python
np.array_equal([1, 2], [1, 2])
```

```python
np.array_equal(np.array([1, 2]), np.array([1, 2]))
```

```python
np.array_equal([1, 2], [1, 2, 3])
```

```python
np.array_equal([1, 2], [1, 4])
```

```python
a = np.array([1, np.nan])
np.array_equal(a, a)
```

```python
np.array_equal(a, a, equal_nan=True)
```

Если ``equal_nan`` равен True, комплексные значения с компонентами nan считаются равными, если nan содержится в вещественной *или* мнимой части.


```python
a = np.array([1 + 1j])
b = a.copy()
a.real = np.nan
b.imag = np.nan
np.array_equal(a, b, equal_nan=True)
```
