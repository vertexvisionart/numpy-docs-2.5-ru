Преобразовать массив углов из радианов в градусы

```python
import numpy as np
rad = np.arange(12.)*np.pi/6
np.degrees(rad)
```

```python
out = np.zeros((rad.shape))
r = np.degrees(rad, out)
np.all(r == out)
```
