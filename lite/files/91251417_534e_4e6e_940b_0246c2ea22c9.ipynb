```python
import numpy as np
def myfunc(a, b):
    "Return a-b if a>b, otherwise return a+b"
    if a > b:
        return a - b
    else:
        return a + b
```

```python
vfunc = np.vectorize(myfunc)
vfunc([1, 2, 3, 4], 2)
```

Если строка документации не задана явно, она берётся из функции, переданной в `vectorize`:


```python
vfunc.__doc__
```

```python
vfunc = np.vectorize(myfunc, doc='Vectorized `myfunc`')
vfunc.__doc__
```

Если тип результата не указан явно, он определяется по результату вычисления
для первого элемента входных данных:


```python
out = vfunc([1, 2, 3, 4], 2)
type(out[0])
```

```python
vfunc = np.vectorize(myfunc, otypes=[float])
out = vfunc([1, 2, 3, 4], 2)
type(out[0])
```

Аргумент `excluded` позволяет исключить некоторые аргументы из векторизации. Это полезно для объектов, подобных массивам, фиксированной длины, например коэффициентов многочлена, как в `polyval`:


```python
def mypolyval(p, x):
    _p = list(p)
    res = _p.pop(0)
    while _p:
        res = res*x + _p.pop(0)
    return res
```

Здесь мы исключаем нулевой аргумент из векторизации, независимо от того, передается ли он по позиции или по ключевому слову.

```python
vpolyval = np.vectorize(mypolyval, excluded={0, 'p'})
vpolyval([1, 2, 3], x=[0, 1])
```

```python
vpolyval(p=[1, 2, 3], x=[0, 1])
```

Аргумент `signature` позволяет векторизовать функции, работающие с нескалярными массивами фиксированной длины. Например, его можно использовать для векторизованного вычисления коэффициента корреляции Пирсона и его p-значения:


```python
import scipy.stats
pearsonr = np.vectorize(scipy.stats.pearsonr,
                signature='(n),(n)->(),()')
pearsonr([[0, 1, 2, 3]], [[1, 2, 3, 4], [4, 3, 2, 1]])
```

Или для векторизованной свертки:

```python
convolve = np.vectorize(np.convolve, signature='(n),(m)->(k)')
convolve(np.eye(4), [1, 2, 1])
```

Поддерживается синтаксис декоратора. Чтобы передать именованные аргументы, декоратор можно вызвать как функцию:


```python
@np.vectorize
def identity(x):
    return x

identity([0, 1, 2])
```

```python
@np.vectorize(otypes=[float])
def as_float(x):
    return x

as_float([0, 1, 2])
```
