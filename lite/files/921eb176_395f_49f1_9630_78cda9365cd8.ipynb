```python
import numpy as np
x = np.arange(16.0).reshape(4, 4)
x
```

```python
np.hsplit(x, 2)
```

```python
np.hsplit(x, np.array([3, 6]))
```

Для массива с большим числом измерений разбиение по-прежнему выполняется вдоль второй оси.


```python
x = np.arange(8.0).reshape(2, 2, 2)
x
```

```python
np.hsplit(x, 2)
```

Одномерный массив разбивается вдоль оси 0.


```python
x = np.array([0, 1, 2, 3, 4, 5])
np.hsplit(x, 2)
```
