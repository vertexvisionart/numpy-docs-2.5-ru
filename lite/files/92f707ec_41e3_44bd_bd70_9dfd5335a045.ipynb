Для вещественных неотрицательных входных значений это работает так же, как `numpy.sqrt`:


```python
import numpy as np
```

```python
np.emath.sqrt(1)
```

```python
np.emath.sqrt([1, 4])
```

Но он автоматически обрабатывает отрицательные входные данные:

```python
np.emath.sqrt(-1)
```

```python
np.emath.sqrt([-1,4])
```

Различные результаты ожидаемы: числа с плавающей точкой 0.0 и -0.0 различаются знаком.

Чтобы явно контролировать результат, используйте complex() следующим образом:

```python
np.emath.sqrt(complex(-4.0, 0.0))
```

```python
np.emath.sqrt(complex(-4.0, -0.0))
```
