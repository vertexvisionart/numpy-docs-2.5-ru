```python
dtype1 = np.dtype([("field1", np.float64), ("field2", np.int64)])
dtype2 = np.dtype([("field1", np.float64)])
np.promote_types(dtype1, dtype2)  # doctest: +IGNORE_EXCEPTION_DETAIL
```
