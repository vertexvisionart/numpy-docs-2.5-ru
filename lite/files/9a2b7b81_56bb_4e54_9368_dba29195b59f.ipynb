Создайте (*очень* грубую) сетку для вычисления множества Мандельброта:

```python
rl = np.linalg.outer(np.ones((5,)), np.linspace(-2, 2, 5))
rl
```

```python
im = np.linalg.outer(1j*np.linspace(2, -2, 5), np.ones((5,)))
im
```

```python
grid = rl + im
grid
```
