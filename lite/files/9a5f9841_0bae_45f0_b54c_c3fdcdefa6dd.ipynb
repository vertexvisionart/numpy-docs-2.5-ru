```python
import numpy as np
a = np.array([[1, 2], [3, 4]])
np.std(a)
```

```python
np.std(a, axis=0)
```

```python
np.std(a, axis=1)
```

При одинарной точности результат std() может быть неточным:


```python
a = np.zeros((2, 512*512), dtype=np.float32)
a[0, :] = 1.0
a[1, :] = 0.1
np.std(a)
```

Вычисление стандартного отклонения в float64 является более точным:

```python
np.std(a, dtype=np.float64)
```

Указание аргумента where:


```python
a = np.array([[14, 8, 11, 10], [7, 9, 10, 11], [10, 15, 5, 10]])
np.std(a)
```

```python
np.std(a, where=[[True], [True], [False]])
```

Использование именованного аргумента mean для экономии времени вычислений:


```python
import numpy as np
from timeit import timeit
a = np.array([[14, 8, 11, 10], [7, 9, 10, 11], [10, 15, 5, 10]])
mean = np.mean(a, axis=1, keepdims=True)

g = globals()
n = 10000
t1 = timeit("std = np.std(a, axis=1, mean=mean)", globals=g, number=n)
t2 = timeit("std = np.std(a, axis=1)", globals=g, number=n)
print(f'Percentage execution time saved {100*(t2-t1)/t2:.0f}%')
```
