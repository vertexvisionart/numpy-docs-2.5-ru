```python
import numpy.ma as ma
x = ma.array([1, 2, 3], mask=[0, 0, 1])
x.mask = True
x
```

Наконец, отдельные элементы можно замаскировать или снять с них маску, присвоив маске последовательность логических значений:


```python
x = ma.array([1, 2, 3])
x.mask = [0, 1, 0]
x
```
