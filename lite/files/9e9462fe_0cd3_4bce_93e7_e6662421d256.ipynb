```python
import numpy as np
data = np.arange(12, dtype=np.float32).reshape((3, 4))
```

В этом примере используется временный файл, так что doctest не записывает файлы в ваш каталог. Вы будете использовать «нормальное» имя файла.

```python
from tempfile import mkdtemp
import os.path as path
filename = path.join(mkdtemp(), 'newfile.dat')
```

Создадим memmap с типом данных dtype и формой, соответствующими нашим данным:


```python
fp = np.memmap(filename, dtype=np.float32, mode='w+', shape=(3,4))
fp
```

Запишите данные в массив memmap:

```python
fp[:] = data[:]
fp
```

```python
fp.filename == path.abspath(filename)
```

Записывает изменения из памяти на диск, чтобы затем прочитать их обратно.

```python
fp.flush()
```

Загрузим memmap и убедимся, что данные сохранены:

```python
newfp = np.memmap(filename, dtype=np.float32, mode='r', shape=(3,4))
newfp
```

memmap только для чтения:

```python
fpr = np.memmap(filename, dtype=np.float32, mode='r', shape=(3,4))
fpr.flags.writeable
```

memmap с копированием при записи:

```python
fpc = np.memmap(filename, dtype=np.float32, mode='c', shape=(3,4))
fpc.flags.writeable
```

В массив с копированием при записи можно присваивать значения, но изменения
затрагивают лишь копию в памяти и не записываются на диск:


```python
fpc
```

```python
fpc[0,:] = 0
fpc
```

Файл на диске не изменился:


```python
fpr
```

Смещение в memmap:


```python
fpo = np.memmap(filename, dtype=np.float32, mode='r', offset=16)
fpo
```
