```python
import numpy as np
from math import pi
np.tan(np.array([-pi,pi/2,pi]))
```

```python

# Example of providing the optional output parameter illustrating
# that what is returned is a reference to said parameter
out1 = np.array([0], dtype=np.float64)
out2 = np.cos([0.1], out1)
out2 is out1
```
