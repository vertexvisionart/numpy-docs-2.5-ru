Для заданной последовательности корней многочлена:

```python
import numpy as np
```

```python
np.poly((0, 0, 0)) # Multiple root example
```

**Комментарии к примеру на русском**

Пример с кратным корнем

Вышеприведенная строка представляет собой z**3 + 0*z**2 + 0*z + 0.

```python
np.poly((-1./2, 0, 1./2))
```

Вышеприведенная строка представляет собой z**3 - z/4

```python
np.poly((np.random.random(1)[0], 0, np.random.random(1)[0]))
```

Для заданного объекта квадратного массива:


```python
P = np.array([[0, 1./3], [-1./2, 0]])
np.poly(P)
```

Обратите внимание: во всех случаях старший коэффициент равен 1.
