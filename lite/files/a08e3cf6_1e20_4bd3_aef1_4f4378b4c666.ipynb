```python
import numpy as np
a = np.array((0, 0, 0, 1, 2, 3, 0, 2, 1, 0))
np.trim_zeros(a)
```

```python
np.trim_zeros(a, trim='b')
```

Поддерживаются несколько измерений.

```python
b = np.array([[0, 0, 2, 3, 0, 0],
              [0, 1, 0, 3, 0, 0],
              [0, 0, 0, 0, 0, 0]])
np.trim_zeros(b)
```

```python
np.trim_zeros(b, axis=-1)
```

Тип входных данных сохраняется: для списка возвращается список, для кортежа — кортеж.


```python
np.trim_zeros([0, 1, 2, 0])
```
