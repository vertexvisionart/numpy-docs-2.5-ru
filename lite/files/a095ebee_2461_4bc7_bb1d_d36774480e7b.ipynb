```python
import numpy as np
```

Вычислим два набора индексов для обращения к массивам 4×4: один для верхней треугольной части, начиная с главной диагонали, а другой — начиная с диагонали на две позиции правее:

```python
iu1 = np.triu_indices(4)
iu1
```

Обратите внимание: индексы строк (первый массив) не убывают, а соответствующие индексы столбцов (второй массив) строго возрастают в пределах каждой строки.

Вот как их можно использовать на примере массива:


```python
a = np.arange(16).reshape(4, 4)
a
```

Оба для индексации:

```python
a[iu1]
```

И для присвоения значений:

```python
a[iu1] = -1
a
```

Они покрывают лишь небольшую часть всего массива (две диагонали справа от основной):

```python
iu2 = np.triu_indices(4, 2)
a[iu2] = -10
a
```
