```python
x = [1e-5, 1e-3, 1e-1]
y = np.arccos(np.cos(x))
np.testing.assert_allclose(x, y, rtol=1e-5, atol=0)
```

Как сказано в примечаниях, `assert_allclose` особым образом обрабатывает скаляры. Здесь тест проверяет, что `numpy.sin` близок к нулю в точках, кратных π.


```python
x = np.arange(3) * np.pi
np.testing.assert_allclose(np.sin(x), 0, atol=1e-15)
```

Используйте `strict`, чтобы при сравнении массива с одним или несколькими измерениями со скаляром возникало исключение ``AssertionError``.


```python
np.testing.assert_allclose(np.sin(x), 0, atol=1e-15, strict=True)
```

Параметр `strict` также гарантирует соответствие типов данных массива:

```python
y = np.zeros(3, dtype=np.float32)
np.testing.assert_allclose(np.sin(x), y, atol=1e-15, strict=True)
```
