```python
import numpy as np
import numpy.ma as ma
a = ma.masked_equal([[1,2],[3,4]], 2)
a
```

```python
ma.getmask(a)
```

Равнозначный способ — использовать у `MaskedArray` атрибут `mask`.


```python
a.mask
```

Результат при mask == `nomask`


```python
b = ma.masked_array([[1,2],[3,4]])
b
```

```python
ma.nomask
```

```python
ma.getmask(b) == ma.nomask
```

```python
b.mask == ma.nomask
```
