```python
from io import StringIO
import numpy as np
```

Файл с разделителями-запятыми и смешанными типами данных


```python
s = StringIO("1,1.3,abcde")
data = np.genfromtxt(s, dtype=[('myint','i8'),('myfloat','f8'),
('mystring','S5')], delimiter=",")
data
```

Использование dtype = None

```python
_ = s.seek(0) # needed for StringIO example only
data = np.genfromtxt(s, dtype=None,
names = ['myint','myfloat','mystring'], delimiter=",")
data
```

**Комментарии к примеру на русском**

Нужно только для примера с StringIO

Задание dtype и имён


```python
_ = s.seek(0)
data = np.genfromtxt(s, dtype="i8,f8,S5",
names=['myint','myfloat','mystring'], delimiter=",")
data
```

Пример с колонками фиксированной ширины

```python
s = StringIO("11.3abcde")
data = np.genfromtxt(s, dtype=None, names=['intvar','fltvar','strvar'],
    delimiter=[1,3,5])
data
```

Пример, чтобы показать комментарии

```python
f = StringIO('''
text,# of chars
hello world,11
numpy,5''')
np.genfromtxt(f, dtype='S12,S12', delimiter=',')
```
