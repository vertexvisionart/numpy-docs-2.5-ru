Определитель двумерного массива [[a, b], [c, d]] равен ad - bc:

```python
import numpy as np
a = np.array([[1, 2], [3, 4]])
np.linalg.det(a)
```

Вычисление определителей для пакета матриц:

```python
a = np.array([ [[1, 2], [3, 4]], [[1, 2], [2, 1]], [[1, 3], [3, 1]] ])
a.shape
```

```python
np.linalg.det(a)
```
