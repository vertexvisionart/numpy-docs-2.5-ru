Производной полинома $x^3 + x^2 + x^1 + 1$ является:

```python
import numpy as np
```

```python
p = np.poly1d([1,1,1,1])
p2 = np.polyder(p)
p2
```

что даёт:

```python
p2(2.)
```

Это можно проверить, приближённо вычислив производную как ``(f(x + h) - f(x))/h``:


```python
(p(2. + 0.001) - p(2.)) / 0.001
```

Производная четвёртого порядка от многочлена третьей степени равна нулю:


```python
np.polyder(p, 2)
```

```python
np.polyder(p, 3)
```

```python
np.polyder(p, 4)
```
