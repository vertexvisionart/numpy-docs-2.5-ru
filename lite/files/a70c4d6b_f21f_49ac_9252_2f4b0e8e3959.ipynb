```python
import numpy as np
np.add(1.0, 4.0)
```

```python
x1 = np.arange(9.0).reshape((3, 3))
x2 = np.arange(3.0)
np.add(x1, x2)
```

Для массивов ndarray оператор ``+`` служит краткой записью ``np.add``.


```python
x1 = np.arange(9.0).reshape((3, 3))
x2 = np.arange(3.0)
x1 + x2
```
