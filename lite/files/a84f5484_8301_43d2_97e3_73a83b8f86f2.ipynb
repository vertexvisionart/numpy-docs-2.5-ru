```python
import numpy as np
from matplotlib.image import NonUniformImage
import matplotlib.pyplot as plt
```

Постройте двумерную гистограмму с переменной шириной интервалов. Сначала задайте их границы:

```python
xedges = [0, 1, 3, 5]
yedges = [0, 2, 3, 4, 6]
```

Далее создаём гистограмму H со случайным содержимым интервалов:

```python
x = np.random.normal(2, 1, 100)
y = np.random.normal(1, 1, 100)
H, xedges, yedges = np.histogram2d(x, y, bins=(xedges, yedges))
# Histogram does not follow Cartesian convention (see Notes),
# therefore transpose H for visualization purposes.
H = H.T
```

**Комментарии к примеру на русском**

Гистограмма не следует декартову соглашению (см. примечания),

поэтому для визуализации транспонируем H.

В приведённом примере `matplotlib.pyplot.imshow` показывает ячейки как квадратные. Примечание переводчика: утверждение оригинала «может отображать только квадратные ячейки» слишком узко. imshow отображает регулярную растровую сетку; размеры ячеек по горизонтали и вертикали можно менять через extent и aspect. Для произвольных неравномерных границ ячеек нужен другой способ отображения:


```python
fig = plt.figure(figsize=(7, 3))
ax = fig.add_subplot(131, title='imshow: square bins')
plt.imshow(H, interpolation='nearest', origin='lower',
        extent=[xedges[0], xedges[-1], yedges[0], yedges[-1]])
```

Функция `matplotlib.pyplot.pcolormesh` может отображать фактические границы ячеек:


```python
ax = fig.add_subplot(132, title='pcolormesh: actual edges',
        aspect='equal')
X, Y = np.meshgrid(xedges, yedges)
ax.pcolormesh(X, Y, H)
```

Класс `matplotlib.image.NonUniformImage` позволяет показывать фактические границы ячеек с интерполяцией:


```python
ax = fig.add_subplot(133, title='NonUniformImage: interpolated',
        aspect='equal', xlim=xedges[[0, -1]], ylim=yedges[[0, -1]])
im = NonUniformImage(ax, interpolation='bilinear')
xcenters = (xedges[:-1] + xedges[1:]) / 2
ycenters = (yedges[:-1] + yedges[1:]) / 2
im.set_data(xcenters, ycenters, H)
ax.add_image(im)
plt.show()
```

Двумерную гистограмму можно построить и без явного задания границ интервалов:

```python
# Generate non-symmetric test data
n = 10000
x = np.linspace(1, 100, n)
y = 2*np.log(x) + np.random.rand(n) - 0.5
# Compute 2d histogram. Note the order of x/y and xedges/yedges
H, yedges, xedges = np.histogram2d(y, x, bins=20)
```

**Комментарии к примеру на русском**

Создаём несимметричные тестовые данные

Вычисляем двумерную гистограмму. Обратите внимание на порядок x/y и xedges/yedges

Теперь построим гистограмму с помощью `matplotlib.pyplot.pcolormesh` и для сравнения используем `matplotlib.pyplot.hexbin`.


```python
# Plot histogram using pcolormesh
fig, (ax1, ax2) = plt.subplots(ncols=2, sharey=True)
ax1.pcolormesh(xedges, yedges, H, cmap='rainbow')
ax1.plot(x, 2*np.log(x), 'k-')
ax1.set_xlim(x.min(), x.max())
ax1.set_ylim(y.min(), y.max())
ax1.set_xlabel('x')
ax1.set_ylabel('y')
ax1.set_title('histogram2d')
ax1.grid()
```

**Комментарии к примеру на русском**

Строим гистограмму с помощью pcolormesh

```python
# Create hexbin plot for comparison
ax2.hexbin(x, y, gridsize=20, cmap='rainbow')
ax2.plot(x, 2*np.log(x), 'k-')
ax2.set_title('hexbin')
ax2.set_xlim(x.min(), x.max())
ax2.set_xlabel('x')
ax2.grid()
```
