```python
from numpy.polynomial.laguerre import lagfromroots, lagval
coef = lagfromroots((-1, 0, 1))
lagval((-1, 0, 1), coef)
```

```python
coef = lagfromroots((-1j, 1j))
lagval((-1j, 1j), coef)
```
