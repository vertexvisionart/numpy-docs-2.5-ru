```python
import numpy as np
```

Число 13 представляется как ``00001101``, а 17 — как ``00010001``. Поэтому побитовое исключающее ИЛИ для 13 и 17 равно ``00011100``, то есть 28:


```python
np.bitwise_xor(13, 17)
```

```python
np.binary_repr(28)
```

```python
np.bitwise_xor(31, 5)
```

```python
np.bitwise_xor([31,3], 5)
```

```python
np.bitwise_xor([31,3], [5,6])
```

```python
np.bitwise_xor([True, True], [False, True])
```

Для массивов ndarray оператор ``@`` служит краткой записью ``np.matmul``.


```python
x1 = np.array([True, True])
x2 = np.array([False, True])
x1 ^ x2
```
