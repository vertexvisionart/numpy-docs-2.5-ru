По умолчанию вычисляется произведение всех элементов:

```python
import numpy as np
np.prod([1.,2.])
```

Даже если входной массив двумерен:

```python
a = np.array([[1., 2.], [3., 4.]])
np.prod(a)
```

Но мы также можем указать ось, по которой умножать:

```python
np.prod(a, axis=1)
```

```python
np.prod(a, axis=0)
```

Или выберем конкретные элементы для включения:


```python
np.prod([1., np.nan, 3.], where=[True, False, True])
```

Если тип `x` беззнаковый, результат имеет платформенный беззнаковый целочисленный тип:


```python
x = np.array([1, 2, 3], dtype=np.uint8)
np.prod(x).dtype == np.uint
```

Если `x` имеет знаковый целочисленный тип, тип результата —
целочисленный тип платформы по умолчанию:


```python
x = np.array([1, 2, 3], dtype=np.int8)
np.prod(x).dtype == int
```

Можно также задать начальное значение произведения, отличное от единицы:


```python
np.prod([1, 2], initial=5)
```
