```python
import numpy as np
np.emath.log(np.exp(1))
```

Отрицательные аргументы обрабатываются «правильно» (напомним, что ``exp(log(x)) == x`` *не выполняется* для вещественных ``x < 0``):


```python
np.emath.log(-np.exp(1)) == (1 + np.pi * 1j)
```
