Создадим равномерную случайную выборку размера 3 из np.arange(5):


```python
rng = np.random.default_rng()
rng.choice(5, 3)
```

```python
#This is equivalent to rng.integers(0,5,3)
```

**Комментарии к примеру на русском**

Эквивалентно rng.integers(0,5,3)

Создадим случайную выборку размера 3 из np.arange(5) с неравными вероятностями выбора элементов:


```python
rng.choice(5, 3, p=[0.1, 0, 0.3, 0.6, 0])
```

Сгенерируем равномерную случайную выборку размера 3 из np.arange(5) без возвращения:


```python
rng.choice(5, 3, replace=False)
```

```python
#This is equivalent to rng.permutation(np.arange(5))[:3]
```

**Комментарии к примеру на русском**

Эквивалентно rng.permutation(np.arange(5))[:3]

Сгенерируйте равномерную случайную выборку без возвращения из двумерного массива вдоль первой оси (по умолчанию):

```python
rng.choice([[0, 1, 2], [3, 4, 5], [6, 7, 8]], 2, replace=False)
```

Сгенерировать выборку из np.arange(5) размером 3 без возвращения, с неравными вероятностями выбора:


```python
rng.choice(5, 3, replace=False, p=[0.1, 0, 0.3, 0.6, 0])
```

Любой из приведённых выше примеров можно повторить с произвольным объектом, подобным массиву, вместо целых чисел. Например:


```python
aa_milne_arr = ['pooh', 'rabbit', 'piglet', 'Christopher']
rng.choice(aa_milne_arr, 5, p=[0.5, 0.1, 0.1, 0.3])
```
