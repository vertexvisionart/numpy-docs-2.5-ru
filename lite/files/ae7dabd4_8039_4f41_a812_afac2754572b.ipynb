```python
a = np.arange(25).reshape(5,5)
b = np.arange(5)
c = np.arange(6).reshape(2,3)
```

```python
np.einsum('ii', a)
```

```python
np.einsum(a, [0,0])
```

```python
np.trace(a)
```

```python
np.einsum('ii->i', a)
```

Сумма по оси (требуется явная форма):

```python
np.einsum('ij->i', a)
```

```python
np.einsum(a, [0,1], [0])
```

```python
np.sum(a, axis=1)
```

Цепочки операций с массивами. Для более сложных свёрток можно ускорить вычисления, каждый раз вычисляя «жадный» порядок либо заранее вычислив «оптимальный» порядок и затем многократно используя его с помощью `einsum_path` (начиная с версии 1.12.0). Для больших массивов прирост производительности может быть особенно значительным:


```python
a = np.ones(64).reshape(2,4,8)
```

Обычный `einsum`: около 1520 мс (измерено на Intel i5 с частотой 3,1 ГГц).

```python
for iteration in range(500):
    _ = np.einsum('ijk,ilm,njm,nlk,abc->',a,a,a,a,a)
```

Неоптимальный `einsum` (из-за повторного вычисления пути): ~330 мс


```python
for iteration in range(500):
    _ = np.einsum('ijk,ilm,njm,nlk,abc->',a,a,a,a,a,
        optimize='optimal')
```

`einsum` с жадным алгоритмом (быстрое приближение оптимального порядка вычислений): около 160 мс


```python
for iteration in range(500):
    _ = np.einsum('ijk,ilm,njm,nlk,abc->',a,a,a,a,a, optimize='greedy')
```

Оптимальный `einsum` (лучший шаблон использования в некоторых случаях): ~ 110 мс

```python
path = np.einsum_path('ijk,ilm,njm,nlk,abc->',a,a,a,a,a,
    optimize='optimal')[0]
for iteration in range(500):
    _ = np.einsum('ijk,ilm,njm,nlk,abc->',a,a,a,a,a, optimize=path)
```
