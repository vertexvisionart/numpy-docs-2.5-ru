```python
import numpy as np
```

Мы уже видели, что 13 представляется как ``00001101``. Его инверсия, то есть побитовое НЕ, равна:


```python
x = np.invert(np.array(13, dtype=np.uint8))
x
```

```python
np.binary_repr(x, width=8)
```

Результат зависит от битовой ширины:

```python
x = np.invert(np.array(13, dtype=np.uint16))
x
```

```python
np.binary_repr(x, width=16)
```

При использовании целых типов со знаком результат равен побитовому НЕ для беззнакового типа, интерпретируемому как целое число со знаком:


```python
np.invert(np.array([13], dtype=np.int8))
```

```python
np.binary_repr(-14, width=8)
```

Логические значения также принимаются:

```python
np.invert(np.array([True, False]))
```

Для массивов ndarray оператор ``@`` служит краткой записью ``np.matmul``.


```python
x1 = np.array([True, False])
~x1
```
