```python
import numpy as np
import numpy.ma as ma
m = [True, False, True, True]
ma.make_mask(m)
```

```python
m = [1, 0, 1, 1]
ma.make_mask(m)
```

```python
m = [1, 0, 2, -3]
ma.make_mask(m)
```

Влияние параметра `shrink`.


```python
m = np.zeros(4)
m
```

```python
ma.make_mask(m)
```

```python
ma.make_mask(m, shrink=False)
```

Использование гибкого `dtype`.


```python
m = [1, 0, 1, 1]
n = [0, 1, 0, 0]
arr = []
for man, mouse in zip(m, n):
    arr.append((man, mouse))
arr
```

```python
dtype = np.dtype({'names':['man', 'mouse'],
                  'formats':[np.int64, np.int64]})
arr = np.array(arr, dtype=dtype)
arr
```

```python
ma.make_mask(arr, dtype=dtype)
```
