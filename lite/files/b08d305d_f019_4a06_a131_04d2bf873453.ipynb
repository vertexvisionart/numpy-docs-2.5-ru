```python
import numpy as np
a = np.arange(24).reshape(2,3,4)
a
```

Суммирование по осям 0 и 2. Результат имеет столько же измерений, сколько исходный массив:


```python
np.apply_over_axes(np.sum, a, [0,2])
```

Кортежи осей в аргументе axis универсальных функций эквивалентны:


```python
np.sum(a, axis=(0,2), keepdims=True)
```
