След вычисляется для двумерных подмассивов, заданных последними двумя осями. Это отличается от `numpy.trace`, которая по умолчанию использует первые две оси.


```python
a = np.arange(24).reshape((3, 2, 2, 2))
np.linalg.trace(a).shape
```

Суммы элементов диагоналей, соседних с главной, можно получить с помощью аргумента `offset`:


```python
a = np.arange(9).reshape((3, 3)); a
```

```python
np.linalg.trace(a, offset=1)  # First superdiagonal
```

**Комментарии к примеру на русском**

Первая диагональ над главной

```python
np.linalg.trace(a, offset=2)  # Second superdiagonal
```

**Комментарии к примеру на русском**

Вторая диагональ над главной

```python
np.linalg.trace(a, offset=-1)  # First subdiagonal
```

**Комментарии к примеру на русском**

Первая диагональ под главной

```python
np.linalg.trace(a, offset=-2)  # Second subdiagonal
```
