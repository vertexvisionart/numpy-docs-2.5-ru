```python
import numpy as np
np.rec.format_parser(['<f8', '<i4'], ['col1', 'col2'],
                     ['T1', 'T2']).dtype
```

`names` и/или `titles` могут быть пустыми списками. Если `titles` является пустым списком, заголовки просто не появятся. Если `names` пуст, будут использоваться имена по умолчанию.

```python
np.rec.format_parser(['f8', 'i4', 'S5'], ['col1', 'col2', 'col3'],
                     []).dtype
```

```python
np.rec.format_parser(['<f8', '<i4', '<S5'], [], []).dtype
```
