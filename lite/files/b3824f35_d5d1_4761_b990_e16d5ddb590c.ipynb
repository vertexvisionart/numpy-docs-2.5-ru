```python
import numpy as np
x = np.ma.MaskedArray([[4, 9, 2, 10],
                       [6, 9, 7, 12]])
```

```python
x.ptp(axis=1)
```

```python
x.ptp(axis=0)
```

```python
x.ptp()
```

Этот пример показывает, что для входного массива знаковых целых чисел может возвращаться отрицательное значение.

```python
y = np.ma.MaskedArray([[1, 127],
                       [0, 127],
                       [-1, 127],
                       [-2, 127]], dtype=np.int8)
y.ptp(axis=1)
```

Обходной путь — использовать метод `view()` для представления результата как беззнаковых целых чисел той же разрядности:


```python
y.ptp(axis=1).view(np.uint8)
```
