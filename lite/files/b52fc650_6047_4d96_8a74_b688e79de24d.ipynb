Для этого массива-примера


```python
a = np.array([[10, 30, 20], [60, 40, 50]])
```

Мы можем заменить максимальные значения на:

```python
ai = np.argmax(a, axis=1, keepdims=True)
ai
```

```python
np.put_along_axis(a, ai, 99, axis=1)
a
```
