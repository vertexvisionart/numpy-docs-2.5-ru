Простая дата в формате ISO:


```python
import numpy as np
```

```python
np.datetime64('2005-02-25')
```

Из целого числа и единицы времени: один год с начала эпохи UNIX:


```python
np.datetime64(1, 'Y')
```

Использование месяцев в качестве единицы измерения:

```python
np.datetime64('2005-02')
```

Укажем только месяц, но явно зададим единицу измерения «дни»:


```python
np.datetime64('2005-02', 'D')
```

Из даты и времени:

```python
np.datetime64('2005-02-25T03:30')
```

NAT (не время):

```python
np.datetime64('nat', 'D')
```

Текущее время (UTC, по умолчанию с точностью до секунды):


```python
np.datetime64('now')
```

```python
np.datetime64('now', 'D')
```

```python
np.datetime64('now', 'ms')
```

Текущая дата:


```python
np.datetime64('today')
```
