np.array позволяет задать непрерывное размещение массива в памяти в порядке C, где быстрее всего меняется последний индекс, или Fortran, где быстрее всего меняется первый индекс.

```python
import numpy as np
a = np.array([[1, 2, 3], [4, 5, 6]], order='C')
a
```

```python
np.isfortran(a)
```

```python
b = np.array([[1, 2, 3], [4, 5, 6]], order='F')
b
```

```python
np.isfortran(b)
```

Транспонирование массива с порядком C даёт массив с порядком Fortran.


```python
a = np.array([[1, 2, 3], [4, 5, 6]], order='C')
a
```

```python
np.isfortran(a)
```

```python
b = a.T
b
```

```python
np.isfortran(b)
```

Для массивов в порядке C проверка даёт False, даже если они одновременно имеют порядок FORTRAN.


```python
np.isfortran(np.array([1, 2], order='F'))
```
