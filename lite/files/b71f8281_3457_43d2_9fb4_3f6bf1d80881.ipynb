```python
import numpy as np
```

```python
a = np.memmap('newfile.dat', dtype=np.float64, mode='w+', shape=1000)
a[10] = 10.0
a[30] = 30.0
del a
```

```python
b = np.fromfile('newfile.dat', dtype=np.float64)
print(b[10], b[30])
```
