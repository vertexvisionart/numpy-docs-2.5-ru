```python
import numpy as np
from numpy import linalg as LA
```

(Почти) тривиальный пример с вещественными собственными значениями и собственными векторами.


```python
eigenvalues, eigenvectors = LA.eig(np.diag((1, 2, 3)))
eigenvalues
```

```python
eigenvectors.real
```

Вещественная матрица с комплексными собственными значениями и собственными векторами. Обратите внимание: собственные значения комплексно сопряжены друг другу.


```python
eigenvalues, eigenvectors = LA.eig(np.array([[1, -1], [1, 1]]))
eigenvalues
```

```python
eigenvectors
```

Комплексная матрица с вещественными собственными значениями и комплексными собственными векторами. Обратите внимание: ``a.conj().T == a``, то есть `a` — эрмитова матрица.


```python
a = np.array([[1, 1j], [-1j, 1]])
eigenvalues, eigenvectors = LA.eig(a)
eigenvalues
```

```python
eigenvectors
```

Будьте осторожны с ошибками округления!

```python
a = np.array([[1 + 1e-9, 0], [0, 1 - 1e-9]])
# Theor. eigenvalues are 1 +/- 1e-9
eigenvalues, eigenvectors = LA.eig(a)
eigenvalues
```
