```python
import numpy as np
```

Число 13 имеет двоичное представление ``00001101``, а 16 — ``00010000``. Поэтому побитовое ИЛИ для 13 и 16 равно ``00011101``, то есть 29:


```python
np.bitwise_or(13, 16)
```

```python
np.binary_repr(29)
```

```python
np.bitwise_or(32, 2)
```

```python
np.bitwise_or([33, 4], 1)
```

```python
np.bitwise_or([33, 4], [1, 2])
```

```python
np.bitwise_or(np.array([2, 5, 255]), np.array([4, 4, 4]))
```

```python
np.array([2, 5, 255]) | np.array([4, 4, 4])
```

```python
np.bitwise_or(np.array([2, 5, 255, 2147483647], dtype=np.int32),
              np.array([4, 4, 4, 2147483647], dtype=np.int32))
```

```python
np.bitwise_or([True, True], [False, True])
```

Для массивов ndarray оператор ``@`` служит краткой записью ``np.matmul``.


```python
x1 = np.array([2, 5, 255])
x2 = np.array([4, 4, 4])
x1 | x2
```
