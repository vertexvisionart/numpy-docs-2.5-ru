```python
import numpy as np
from tempfile import TemporaryFile
outfile = TemporaryFile()
x = np.arange(10)
y = np.sin(x)
```

При использовании `savez` с \*args массивы сохраняются с именами по умолчанию.

```python
np.savez(outfile, x, y)
_ = outfile.seek(0) # Only needed to simulate closing & reopening file
npzfile = np.load(outfile)
npzfile.files
```

**Комментарии к примеру на русском**

Нужно лишь для имитации закрытия и повторного открытия файла

```python
npzfile['arr_0']
```

Используя `savez` с \**kwds, массивы сохраняются с именами ключевых слов.

```python
outfile = TemporaryFile()
np.savez(outfile, x=x, y=y)
_ = outfile.seek(0)
npzfile = np.load(outfile)
sorted(npzfile.files)
```
