```python
import numpy as np
import numpy.ma as ma
x = np.array([1, 1.1, 2, 1.1, 3])
ma.masked_values(x, 1.1)
```

Обратите внимание: `mask` по возможности задаётся равным ``nomask``.


```python
ma.masked_values(x, 2.1)
```

В отличие от `masked_equal`, `masked_values` позволяет проверять приближённое равенство.


```python
ma.masked_values(x, 2.1, atol=1e-1)
```
