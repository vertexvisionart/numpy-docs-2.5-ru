Для массивов, содержащих объекты Python (например, dtype=np.object_),
создаётся поверхностная копия. Новый массив содержит те же объекты,
что может привести к неожиданным результатам, если эти объекты
допускают изменение:


```python
a = np.array([1, 'm', [2, 3, 4]], dtype=np.object_)
b = a.copy()
b[2][0] = 10
a
```

Чтобы гарантировать копирование всех элементов массива типа ``object``, используйте `copy.deepcopy`:


```python
import copy
a = np.array([1, 'm', [2, 3, 4]], dtype=np.object_)
c = copy.deepcopy(a)
c[2][0] = 10
c
```
