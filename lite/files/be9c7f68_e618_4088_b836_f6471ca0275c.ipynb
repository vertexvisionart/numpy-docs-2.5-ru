```python
import numpy as np
```

Начнём с массива целых чисел от 0 до 5 включительно. Элементы меньше ``3`` меняют знак, элементы больше ``3`` возводятся в квадрат, а элементы, не удовлетворяющие ни одному условию (то есть равные ``3``), заменяются значением `default`, равным ``42``.


```python
x = np.arange(6)
condlist = [x<3, x>3]
choicelist = [-x, x**2]
np.select(condlist, choicelist, 42)
```

При выполнении нескольких условий используется первое, встречающееся в `condlist`.

```python
condlist = [x<=4, x>3]
choicelist = [x, x**2]
np.select(condlist, choicelist, 55)
```
