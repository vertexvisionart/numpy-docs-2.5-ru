```python
import numpy as np
a = np.array([[1, 2], [3, 4]])
b = np.array([[5, 6]])
np.concatenate((a, b), axis=0)
```

```python
np.concatenate((a, b.T), axis=1)
```

```python
np.concatenate((a, b), axis=None)
```

Эта функция не будет сохранять маскировку входов MaskedArray.

```python
a = np.ma.arange(3)
a[1] = np.ma.masked
b = np.arange(2, 5)
a
```

```python
b
```

```python
np.concatenate([a, b])
```

```python
np.ma.concatenate([a, b])
```
