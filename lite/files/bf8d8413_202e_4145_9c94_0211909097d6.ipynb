```python
import numpy as np
dt = np.dtype(np.int16)
dt.byteorder
```

```python
# endian is not relevant for 8 bit numbers
np.dtype(np.int8).byteorder
```

**Комментарии к примеру на русском**

Для 8-битных чисел порядок байтов не имеет значения

```python
# or ASCII strings
np.dtype('S2').byteorder
```

**Комментарии к примеру на русском**

Как и для строк ASCII

```python
# Even if specific code is given, and it is native
# '=' is the byteorder
import sys
sys_is_le = sys.byteorder == 'little'
native_code = '<' if sys_is_le else '>'
swapped_code = '>' if sys_is_le else '<'
dt = np.dtype(native_code + 'i2')
dt.byteorder
```

**Комментарии к примеру на русском**

Даже если явно задан код машинного порядка байтов,

byteorder будет равен «=».

```python
# Swapped code shows up as itself
dt = np.dtype(swapped_code + 'i2')
dt.byteorder == swapped_code
```
