Следующее утверждение проходит, потому что каждый конечный элемент `x` строго меньше соответствующего элемента `y`, а NaN находятся в соответствующих местах.

```python
x = [1.0, 1.0, np.nan]
y = [1.1, 2.0, np.nan]
np.testing.assert_array_less(x, y)
```

Следующая проверка не проходит, поскольку нулевой элемент `x` больше не строго меньше нулевого элемента `y`.


```python
y[0] = 1
np.testing.assert_array_less(x, y)
```

Здесь `y` — скаляр, поэтому каждый элемент `x` сравнивается с `y`, и проверка проходит.

```python
x = [1.0, 4.0]
y = 5.0
np.testing.assert_array_less(x, y)
```

Однако при ``strict=True`` проверка завершится ошибкой из-за несовпадения форм.


```python
np.testing.assert_array_less(x, y, strict=True)
```

При ``strict=True`` проверка также завершается ошибкой, если типы данных двух массивов не совпадают.


```python
y = [5, 5]
np.testing.assert_array_less(x, y, strict=True)
```
