Сгенерируем выборку из распределения:


```python
a = 4.0
n = 20000
s = np.random.zipf(a, n)
```

Построим гистограмму выборки вместе с ожидаемой гистограммой, рассчитанной по функции плотности вероятности:


```python
import matplotlib.pyplot as plt
from scipy.special import zeta  # doctest: +SKIP
```

**Комментарии к примеру на русском**

doctest: +SKIP

`bincount` позволяет быстро построить гистограмму для небольших целых чисел.


```python
count = np.bincount(s)
k = np.arange(1, s.max() + 1)
```

```python
plt.bar(k, count[1:], alpha=0.5, label='sample count')
plt.plot(k, n*(k**-a)/zeta(a), 'k.-', alpha=0.5,
         label='expected count')   # doctest: +SKIP
plt.semilogy()
plt.grid(alpha=0.4)
plt.legend()
plt.title(f'Zipf sample, a={a}, size={n}')
plt.show()
```
