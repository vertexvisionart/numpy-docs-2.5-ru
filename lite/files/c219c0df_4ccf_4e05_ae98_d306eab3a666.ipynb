```python
import numpy as np
np.fft.fft(np.exp(2j * np.pi * np.arange(8) / 8))
```

В этом примере БПФ вещественных входных данных обладает эрмитовой симметрией: вещественная часть симметрична, а мнимая — антисимметрична, как описано в документации `numpy.fft`:

```python
import matplotlib.pyplot as plt
t = np.arange(256)
sp = np.fft.fft(np.sin(t))
freq = np.fft.fftfreq(t.shape[-1])
_ = plt.plot(freq, sp.real, freq, sp.imag)
plt.show()
```
