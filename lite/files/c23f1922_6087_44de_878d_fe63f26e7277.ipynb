В этом примере создадим два случайных массива, ``xarr`` и ``yarr``, и вычислим коэффициенты корреляции Пирсона ``R`` по строкам и столбцам. Поскольку ``rowvar`` по умолчанию имеет истинное значение, сначала найдём корреляции между переменными, представленными строками ``xarr``.


```python
import numpy as np
rng = np.random.default_rng(seed=42)
xarr = rng.random((3, 3))
xarr
```

```python
R1 = np.corrcoef(xarr)
R1
```

Добавив ещё один набор переменных и наблюдений ``yarr``, можно вычислить по строкам коэффициенты корреляции Пирсона между переменными в ``xarr`` и ``yarr``.


```python
yarr = rng.random((3, 3))
yarr
```

```python
R2 = np.corrcoef(xarr, yarr)
R2
```

Наконец, при ``rowvar=False`` переменными считаются столбцы, и мы найдём коэффициенты корреляции Пирсона между переменными, представленными столбцами ``xarr`` и ``yarr``.


```python
R3 = np.corrcoef(xarr, yarr, rowvar=False)
R3
```
