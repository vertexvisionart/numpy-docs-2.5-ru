Сгенерируем выборку из распределения:


```python
rng = np.random.default_rng()
lam, size = 5, 10000
s = rng.poisson(lam=lam, size=size)
```

Проверим среднее и дисперсию: оба значения должны быть приблизительно равны ``lam``:


```python
s.mean(), s.var()
```

Построим гистограмму и функцию вероятностей:


```python
import matplotlib.pyplot as plt
from scipy import stats
x = np.arange(0, 21)
pmf = stats.poisson.pmf(x, mu=lam)
plt.hist(s, bins=x, density=True, width=0.5)
plt.stem(x, pmf, 'C1-')
plt.show()
```
