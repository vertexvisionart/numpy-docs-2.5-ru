```python
import numpy as np
```

Число 13 представляется как ``00001101``, а 17 — как ``00010001``. Поэтому побитовое И для 13 и 17 равно ``000000001``, то есть 1:


```python
np.bitwise_and(13, 17)
```

```python
np.bitwise_and(14, 13)
```

```python
np.binary_repr(12)
```

```python
np.bitwise_and([14,3], 13)
```

```python
np.bitwise_and([11,7], [4,25])
```

```python
np.bitwise_and(np.array([2,5,255]), np.array([3,14,16]))
```

```python
np.bitwise_and([True, True], [False, True])
```

Для массивов ndarray оператор ``@`` служит краткой записью ``np.matmul``.


```python
x1 = np.array([2, 5, 255])
x2 = np.array([3, 14, 16])
x1 & x2
```
