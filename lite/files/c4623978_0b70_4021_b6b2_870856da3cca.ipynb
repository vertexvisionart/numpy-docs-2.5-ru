Начнём с массива, непрерывно расположенного в памяти в порядке C:


```python
import numpy as np
x = np.ones((2, 3), order='C')
x.flags['C_CONTIGUOUS']
```

Вызов ``asfortranarray`` создаёт копию, непрерывно расположенную в памяти в порядке Fortran:


```python
y = np.asfortranarray(x)
y.flags['F_CONTIGUOUS']
```

```python
np.may_share_memory(x, y)
```

Теперь начнём с массива, непрерывно расположенного в памяти в порядке Fortran:


```python
x = np.ones((2, 3), order='F')
x.flags['F_CONTIGUOUS']
```

Тогда вызов ``asfortranarray`` возвращает тот же объект:


```python
y = np.asfortranarray(x)
x is y
```

Примечание: функция возвращает массив как минимум с одним измерением, поэтому нульмерные массивы не сохраняются.
