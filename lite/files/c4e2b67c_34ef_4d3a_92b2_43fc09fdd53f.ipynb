```python
import numpy as np
np.geomspace(1, 1000, num=4)
```

```python
np.geomspace(1, 1000, num=3, endpoint=False)
```

```python
np.geomspace(1, 1000, num=4, endpoint=False)
```

```python
np.geomspace(1, 256, num=9)
```

Обратите внимание: приведённый выше код может давать значения, не являющиеся точными целыми числами:


```python
np.geomspace(1, 256, num=9, dtype=np.int_)
```

```python
np.around(np.geomspace(1, 256, num=9)).astype(np.int_)
```

Допускаются отрицательные, убывающие и комплексные входные значения:


```python
np.geomspace(1000, 1, num=4)
```

```python
np.geomspace(-1000, -1, num=4)
```

```python
np.geomspace(1j, 1000j, num=4)  # Straight line
```

**Комментарии к примеру на русском**

Прямая линия

```python
np.geomspace(-1+0j, 1+0j, num=5)  # Circle
```

Графическое изображение параметра `endpoint`:

```python
import matplotlib.pyplot as plt
N = 10
y = np.zeros(N)
plt.semilogx(np.geomspace(1, 1000, N, endpoint=True), y + 1, 'o')
```

```python
plt.semilogx(np.geomspace(1, 1000, N, endpoint=False), y + 2, 'o')
```

```python
plt.axis([0.5, 2000, 0, 3])
```

```python
plt.grid(True, color='0.7', linestyle='-', which='both', axis='both')
plt.show()
```
