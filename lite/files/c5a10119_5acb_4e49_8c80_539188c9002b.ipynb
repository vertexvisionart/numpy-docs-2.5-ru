```python
import numpy as np
f = np.array([1, 2, 4, 7, 11, 16])
np.gradient(f)
```

```python
np.gradient(f, 2)
```

Шаги можно также задать массивом координат значений F вдоль измерений. Например, для равномерной сетки:


```python
x = np.arange(f.size)
np.gradient(f, x)
```

Или неравномерный:


```python
x = np.array([0., 1., 1.5, 3.5, 4., 6.])
np.gradient(f, x)
```

Для двумерного массива возвращаются два массива в порядке осей.
В этом примере первый массив содержит градиент
в направлении строк, а второй — в направлении столбцов:


```python
np.gradient(np.array([[1, 2, 6], [3, 4, 5]]))
```

В этом примере также задан шаг: равномерный для axis=0 и неравномерный для axis=1.


```python
dx = 2.
y = [1., 1.5, 3.5]
np.gradient(np.array([[1, 2, 6], [3, 4, 5]]), dx, y)
```

Способ обработки границ можно задать с помощью `edge_order`.


```python
x = np.array([0, 1, 2, 3, 4])
f = x**2
np.gradient(f, edge_order=1)
```

```python
np.gradient(f, edge_order=2)
```

Именованный аргумент `axis` позволяет выбрать подмножество осей, по которым вычисляется градиент.

```python
np.gradient(np.array([[1, 2, 6], [3, 4, 5]]), axis=0)
```

Аргумент `varargs` задаёт расстояния между точками входного массива. Возможны две формы:

1. Массив координат; они могут быть неравномерно расположены:


```python
x = np.array([0., 2., 3., 6., 8.])
y = x ** 2
np.gradient(y, x, edge_order=2)
```

2. Скаляр, задающий постоянное расстояние между отсчётами:


```python
dx = 2
x = np.array([0., 2., 4., 6., 8.])
y = x ** 2
np.gradient(y, dx, edge_order=2)
```

Шаги можно задать отдельно для каждого измерения. Число аргументов должно совпадать с числом измерений входных данных.

```python
dx = 2
dy = 3
x = np.arange(0, 6, dx)
y = np.arange(0, 9, dy)
xs, ys = np.meshgrid(x, y)
zs = xs + 2 * ys
np.gradient(zs, dy, dx)  # Passing two scalars
```

Допускается также смешивание скаляров и массивов:

```python
np.gradient(zs, y, dx)  # Passing one array and one scalar
```
