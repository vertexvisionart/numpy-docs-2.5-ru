```python
import numpy as np
x = np.array([1e-16,1,2,3])
np.array2string(x, precision=2, separator=',',
                      suppress_small=True)
```

```python
x  = np.arange(3.)
np.array2string(x, formatter={'float_kind':lambda x: "%.2f" % x})
```

```python
x  = np.arange(3)
np.array2string(x, formatter={'int':lambda x: hex(x)})
```
