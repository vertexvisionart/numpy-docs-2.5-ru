```python
import numpy as np
np.append([1, 2, 3], [[4, 5, 6], [7, 8, 9]])
```

Если указан `axis`, `values` должен иметь правильную форму.


```python
np.append([[1, 2, 3], [4, 5, 6]], [[7, 8, 9]], axis=0)
```

```python
np.append([[1, 2, 3], [4, 5, 6]], [7, 8, 9], axis=0)
```

```python
a = np.array([1, 2], dtype=np.int_)
c = np.append(a, [])
c
```

```python
c.dtype
```

Тип данных пустого ndarray по умолчанию — `float64`, поэтому результат имеет dtype `float64` и при добавлении данных типа `int64`.
