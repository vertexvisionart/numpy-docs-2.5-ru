```python
import numpy as np
```

Возведём каждый элемент списка в куб.

```python
x1 = range(6)
x1
```

```python
np.float_power(x1, 3)
```

Возводим основания в разные степени.

```python
x2 = [1.0, 2.0, 3.0, 3.0, 2.0, 1.0]
np.float_power(x1, x2)
```

Влияние трансляции массивов.

```python
x2 = np.array([[1, 2, 3, 3, 2, 1], [1, 2, 3, 3, 2, 1]])
x2
```

```python
np.float_power(x1, x2)
```

Возведение отрицательных значений в нецелую степень даёт ``nan`` и сопровождается предупреждением.


```python
x3 = np.array([-1, -4])
with np.errstate(invalid='ignore'):
    p = np.float_power(x3, 1.5)

p
```

Чтобы получить комплексный результат, передайте аргумент ``dtype=np.complex128``.


```python
np.float_power(x3, 1.5, dtype=np.complex128)
```
