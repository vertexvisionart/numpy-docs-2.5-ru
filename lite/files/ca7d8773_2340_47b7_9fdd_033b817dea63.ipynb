Простой пример корреляции:


```python
a = np.ma.array([1, 2, 3])
v = np.ma.array([0, 1, 0])
np.ma.correlate(a, v, mode='valid')
```

Корреляция с замаскированными элементами:


```python
a = np.ma.array([1, 2, 3], mask=[False, True, False])
v = np.ma.array([0, 1, 0])
np.ma.correlate(a, v, mode='valid', propagate_mask=True)
```

Корреляция в разных режимах и для массивов разных типов:


```python
a = np.ma.array([1, 2, 3])
v = np.ma.array([0, 1, 0])
np.ma.correlate(a, v, mode='full')
```
