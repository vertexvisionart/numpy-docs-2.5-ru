```python
import numpy as np
arr = np.arange(12).reshape((3, 4))
arr
```

```python
condition = np.mod(arr, 3)==0
condition
```

```python
np.extract(condition, arr)
```

Если `condition` является булевым:

```python
arr[condition]
```
