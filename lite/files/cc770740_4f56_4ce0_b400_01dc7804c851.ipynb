```python
import numpy as np
from numpy.lib.stride_tricks import as_strided
x = np.arange(10)
y = as_strided(x, shape=(5,), strides=(8,), check_bounds=True)
y
```

Попытка создать представление за границами массива с ``check_bounds=True`` в as_strided вызовет ошибку:


```python
as_strided(x, shape=(20,), strides=(8,), check_bounds=True)
```

При работе с представлениями границы проверяются относительно базового массива:


```python
a = np.arange(1000)
b = a[:2]
c = as_strided(b, shape=(2,), strides=(400,), check_bounds=True)
c[0], c[1]
```
