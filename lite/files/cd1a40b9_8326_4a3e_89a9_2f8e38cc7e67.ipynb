```python
import numpy as np
np.round([0.37, 1.64])
```

```python
np.round([0.37, 1.64], decimals=1)
```

```python
np.round([.5, 1.5, 2.5, 3.5, 4.5]) # rounds to nearest even value
```

**Комментарии к примеру на русском**

Округляет к ближайшему чётному значению

```python
np.round([1,2,3,11], decimals=1) # ndarray of ints is returned
```

**Комментарии к примеру на русском**

Возвращается ndarray целых чисел

```python
np.round([1,2,3,11], decimals=-1)
```
