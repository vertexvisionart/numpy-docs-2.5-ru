Матрица Вандермонда степени ``deg = 5`` для точек ``x = [-1, 2, 3]`` содержит в столбцах поэлементные степени `x` от 0 до 5.


```python
from numpy.polynomial import polynomial as P
x, deg = [-1, 2, 3], 5
P.polyvander(x=x, deg=deg)
```
