Генерируем десять тысяч значений из геометрического распределения с вероятностью успеха в отдельном испытании 0,35:

```python
z = np.random.geometric(p=0.35, size=10000)
```

Сколько испытаний завершилось успехом за один запуск?

```python
(z == 1).sum() / 10000.
```
