```python
import numpy as np
A = np.asmatrix('1 1; 1 1')
B = np.asmatrix('2 2; 2 2')
C = np.asmatrix('3 4; 5 6')
D = np.asmatrix('7 8; 9 0')
```

Все следующие выражения создают одну и ту же блочную матрицу:

```python
np.bmat([[A, B], [C, D]])
```

```python
np.bmat(np.r_[np.c_[A, B], np.c_[C, D]])
```

```python
np.bmat('A,B; C,D')
```
