Изменить знак элементов 0 и 1:


```python
import numpy as np
a = np.array([1, 2, 3, 4])
np.negative.at(a, [0, 1])
a
```

Увеличиваем элементы 0 и 1, а элемент 2 — дважды:

```python
a = np.array([1, 2, 3, 4])
np.add.at(a, [0, 1, 2, 2], 1)
a
```

Добавьте элементы 0 и 1 в первом массиве ко второму массиву и сохраните результаты в первом массиве:

```python
a = np.array([1, 2, 3, 4])
b = np.array([1, 2])
np.add.at(a, [0, 1], b)
a
```
