```python
import numpy as np
np.nancumsum(1)
```

```python
np.nancumsum([1])
```

```python
np.nancumsum([1, np.nan])
```

```python
a = np.array([[1, 2], [3, np.nan]])
np.nancumsum(a)
```

```python
np.nancumsum(a, axis=0)
```

```python
np.nancumsum(a, axis=1)
```
