```python
import numpy as np
np.nansum(1)
```

```python
np.nansum([1])
```

```python
np.nansum([1, np.nan])
```

```python
a = np.array([[1, 1], [1, np.nan]])
np.nansum(a)
```

```python
np.nansum(a, axis=0)
```

```python
np.nansum([1, np.nan, np.inf])
```

```python
np.nansum([1, np.nan, -np.inf])
```

```python
with np.errstate(invalid="ignore"):
    np.nansum([1, np.nan, np.inf, -np.inf]) # both +/- infinity present
```
