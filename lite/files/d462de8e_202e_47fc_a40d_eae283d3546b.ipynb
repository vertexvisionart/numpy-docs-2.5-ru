Используя тот факт, что собственные значения диагональной матрицы равны её диагональным элементам, покажем: умножение матрицы слева на ортогональную матрицу `Q`, а справа на `Q.T` (транспонированную `Q`) сохраняет собственные значения «средней» матрицы. Иными словами, если `Q` ортогональна, то ``Q * A * Q.T`` имеет те же собственные значения, что и ``A``:


```python
import numpy as np
from numpy import linalg as LA
x = np.random.random()
Q = np.array([[np.cos(x), -np.sin(x)], [np.sin(x), np.cos(x)]])
LA.norm(Q[0, :]), LA.norm(Q[1, :]), np.dot(Q[0, :],Q[1, :])
```

Теперь умножим диагональную матрицу на ``Q`` с одной стороны и на ``Q.T`` с другой:


```python
D = np.diag((-1,1))
LA.eigvals(D)
```

```python
A = np.dot(Q, D)
A = np.dot(A, Q.T)
LA.eigvals(A)
```
