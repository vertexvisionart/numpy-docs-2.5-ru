Векторное произведение.

```python
x = np.array([1, 2, 3])
y = np.array([4, 5, 6])
np.linalg.cross(x, y)
```

Несколько векторных произведений. Направление вектора результата определяется *правилом правой руки*.

```python
x = np.array([[1,2,3], [4,5,6]])
y = np.array([[4,5,6], [1,2,3]])
np.linalg.cross(x, y)
```

```python
x = np.array([[1, 2], [3, 4], [5, 6]])
y = np.array([[4, 5], [6, 1], [2, 3]])
np.linalg.cross(x, y, axis=0)
```
