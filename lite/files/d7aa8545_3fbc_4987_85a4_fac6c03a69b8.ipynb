```python
import numpy as np
from numpy.linalg import matrix_power
i = np.array([[0, 1], [-1, 0]]) # matrix equiv. of the imaginary unit
matrix_power(i, 3) # should = -i
```

**Комментарии к примеру на русском**

Матричный аналог мнимой единицы

Должно быть равно -i

```python
matrix_power(i, 0)
```

```python
matrix_power(i, -3) # should = 1/(-i) = i, but w/ f.p. elements
```

**Комментарии к примеру на русском**

Должно быть 1/(-i) = i, но элементы с плавающей точкой

Несколько более сложный пример

```python
q = np.zeros((4, 4))
q[0:2, 0:2] = -i
q[2:4, 2:4] = i
q # one of the three quaternion units not equal to 1
```

**Комментарии к примеру на русском**

Одна из трёх кватернионных единиц, не равных 1

```python
matrix_power(q, 2) # = -np.eye(4)
```

**Комментарии к примеру на русском**

= -np.eye(4)
