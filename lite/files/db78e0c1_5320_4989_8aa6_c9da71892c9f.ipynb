В качестве иллюстрации рассмотрим следующий набор данных:

````python
import numpy as np
import numpy.ma as ma
x = np.array([1, 2, 3, -1, 5])
````

Нужно пометить четвёртый элемент как недопустимый. Проще всего создать массив с маской
```

```

````python
mx = ma.masked_array(x, mask=[0, 0, 0, 1, 0])
````

Теперь можно вычислить среднее значение набора данных, не учитывая недопустимые данные:


````python
mx.mean()
````
