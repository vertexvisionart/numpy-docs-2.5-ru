32-битное целое число, первые два байта которого интерпретируются как целое число через поле ``real``, а следующие два — через поле ``imag``.


```python
import numpy as np
```

```python
dt = np.dtype((np.int32,{'real':(np.int16, 0),'imag':(np.int16, 2)}))
```

32-битное целое число, интерпретируемое как подмассив формы ``(4,)`` из 8-битных целых чисел:


```python
dt = np.dtype((np.int32, (np.int8, 4)))
```

32-битное целое число с полями ``r``, ``g``, ``b``, ``a``, интерпретирующими его четыре байта как четыре беззнаковых целых числа:


```python
dt = np.dtype(('i4', [('r','u1'),('g','u1'),('b','u1'),('a','u1')]))
```
