```python
import numpy as np
from tempfile import TemporaryFile
outfile = TemporaryFile()
x = np.arange(10)
y = np.sin(x)
np.savez(outfile, x=x, y=y)
_ = outfile.seek(0)
```

```python
npz = np.load(outfile)
isinstance(npz, np.lib.npyio.NpzFile)
```

```python
npz
```

```python
sorted(npz.files)
```

```python
npz['x']  # getitem access
```
