```python
import numpy as np
a = np.arange(3 * 4 * 5 * 6).reshape(3, 4, 5, 6)
a_itor = np.lib.Arrayterator(a, 2)
a_itor.shape
```

Теперь можно обходить ``a_itor``, получая массивы размера два. Поскольку `buf_size` меньше размера любого измерения, сначала будет обходиться первое измерение:


```python
for subarr in a_itor:
    if not subarr.all():
        print(subarr, subarr.shape) # doctest: +SKIP
# [[[[0 1]]]] (1, 1, 1, 2)
```
