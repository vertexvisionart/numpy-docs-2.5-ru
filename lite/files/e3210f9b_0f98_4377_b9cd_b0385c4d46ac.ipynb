```python
import numpy as np
a = np.arange(6).reshape(2,3) + 10
a
```

```python
np.argmin(a)
```

```python
np.argmin(a, axis=0)
```

```python
np.argmin(a, axis=1)
```

Индексы минимальных элементов N-мерного массива:

```python
a.flat[np.argmin(a)]
```

```python
ind = np.unravel_index(np.argmin(a, axis=None), a.shape)
ind
```

```python
a[ind]
```

```python
b = np.arange(6) + 10
b[4] = 10
b
```

```python
np.argmin(b)  # Only the first occurrence is returned.
```

**Комментарии к примеру на русском**

Возвращается только первое вхождение.

```python
x = np.array([[4,2,3], [1,0,3]])
index_array = np.argmin(x, axis=-1)
# Same as np.amin(x, axis=-1, keepdims=True)
np.take_along_axis(x, np.expand_dims(index_array, axis=-1), axis=-1)
```

**Комментарии к примеру на русском**

То же, что np.amin(x, axis=-1, keepdims=True)

```python
# Same as np.amax(x, axis=-1)
np.take_along_axis(x, np.expand_dims(index_array, axis=-1),
    axis=-1).squeeze(axis=-1)
```

**Комментарии к примеру на русском**

То же, что np.amax(x, axis=-1)

Если задать `keepdims` равным `True`,


```python
x = np.arange(24).reshape((2, 3, 4))
res = np.argmin(x, axis=1, keepdims=True)
res.shape
```
