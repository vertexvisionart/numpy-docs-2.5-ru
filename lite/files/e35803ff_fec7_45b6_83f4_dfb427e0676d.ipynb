```python
import numpy as np
```

Создайте массив x, со ссылкой y и копией z:

```python
x = np.array([1, 2, 3])
y = x
z = np.copy(x)
```

Обратите внимание: при изменении x меняется y, но не z:


```python
x[0] = 10
x[0] == y[0]
```

```python
x[0] == z[0]
```

Обратите внимание: np.copy сбрасывает ранее установленный флаг WRITEABLE=False.

```python
a = np.array([1, 2, 3])
a.flags["WRITEABLE"] = False
b = np.copy(a)
b.flags["WRITEABLE"]
```

```python
b[0] = 3
b
```
