Пример из Glantz [1], стр. 47–40:

Рассмотрим две группы: дети людей с диабетом (25 человек) и дети людей без диабета (контрольная группа, 25 человек). Измеряли уровень глюкозы в крови натощак. В первой группе среднее значение составило 86,1, в контрольной — 82,2; стандартные отклонения — 2,09 и 2,49 соответственно. Согласуются ли эти данные с нулевой гипотезой о том, что наличие диабета у родителей не влияет на уровень глюкозы в крови их детей? Вычисленная по данным F-статистика равна 36,01.

Сгенерируем выборку из распределения:


```python
dfnum = 1. # between group degrees of freedom
dfden = 48. # within groups degrees of freedom
rng = np.random.default_rng()
s = rng.f(dfnum, dfden, 1000)
```

```python
np.sort(s)[-10]
```

Итак, вероятность того, что F-статистика превысит 7,62, составляет около 1%. Измеренное значение равно 36, поэтому нулевая гипотеза отвергается на уровне значимости 1%.

Соответствующая функция плотности вероятности при ``n = 20`` и ``m = 20``:


```python
import matplotlib.pyplot as plt
from scipy import stats
dfnum, dfden, size = 20, 20, 10000
s = rng.f(dfnum=dfnum, dfden=dfden, size=size)
bins, density, _ = plt.hist(s, 30, density=True)
x = np.linspace(0, 5, 1000)
plt.plot(x, stats.f.pdf(x, dfnum, dfden))
plt.xlim([0, 5])
plt.show()
```
