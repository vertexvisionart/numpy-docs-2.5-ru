```python
import numpy as np
from numpy.lib import recfunctions as rfn
a = np.array([(1, (2, 3.0)), (4, (5, 6.0))],
  dtype=[('a', np.int64), ('b', [('ba', np.double), ('bb', np.int64)])])
rfn.drop_fields(a, 'a')
```

```python
rfn.drop_fields(a, 'ba')
```

```python
rfn.drop_fields(a, ['ba', 'bb'])
```
