Рассмотрим четыре точки в разных квадрантах:


```python
import numpy as np
x = np.array([-1, +1, +1, -1])
y = np.array([-1, -1, +1, +1])
np.arctan2(y, x) * 180 / np.pi
```

Обратите внимание на порядок параметров. `arctan2` определена и при `x2` = 0, а также в ряде других особых точек; её значения лежат в диапазоне ``[-pi, pi]``:


```python
np.arctan2([1., -1.], [0., 0.])
```

```python
np.arctan2([0., 0., np.inf], [+0., -0., np.inf])
```
