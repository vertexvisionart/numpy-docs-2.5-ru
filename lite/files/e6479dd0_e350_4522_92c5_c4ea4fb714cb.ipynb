Основной пример использования. Обратите внимание: y — «развёрнутая» версия [a[:, 0, :], a[:, 1, 0], a[:, 2, :]], поскольку оси первого итератора заданы как [1].

```python
import numpy as np
a = np.arange(12).reshape(2, 3, 2)
i, j = np.nested_iters(a, [[1], [0, 2]], flags=["multi_index"])
for x in i:
     print(i.multi_index)
     for y in j:
         print('', j.multi_index, y)
```
