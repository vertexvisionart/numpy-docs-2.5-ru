```python
import numpy as np
a = np.array([[1, 2], [3, 4]])
a
```

```python
np.transpose(a)
```

```python
a = np.array([1, 2, 3, 4])
a
```

```python
np.transpose(a)
```

```python
a = np.ones((1, 2, 3))
np.transpose(a, (1, 0, 2)).shape
```

```python
a = np.ones((2, 3, 4, 5))
np.transpose(a).shape
```

```python
a = np.arange(3*4*5).reshape((3, 4, 5))
np.transpose(a, (-1, 0, -2)).shape
```
