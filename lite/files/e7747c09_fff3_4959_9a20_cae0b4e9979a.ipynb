Начнём с генератора по умолчанию с заданным начальным значением:


```python
# High quality entropy created with: f"0x{secrets.randbits(128):x}"
entropy = 0x3034c61a9ae04ff8cb62ab8ec2c4b501
rng = np.random.default_rng(entropy)
```

**Комментарии к примеру на русском**

Высококачественная энтропия получена с помощью f"0x{secrets.randbits(128):x}"

Создать два новых генератора для параллельного выполнения:

```python
child_rng1, child_rng2 = rng.spawn(2)
```

Числа, генерируемые каждым из них, независимы, но происходят из одной исходной энтропии инициализации:


```python
rng.uniform(), child_rng1.uniform(), child_rng2.uniform()
```

Можно безопасно создавать дополнительные дочерние генераторы как из исходного ``rng``, так и из дочерних генераторов:


```python
more_child_rngs = rng.spawn(20)
nested_spawn = child_rng1.spawn(20)
```
