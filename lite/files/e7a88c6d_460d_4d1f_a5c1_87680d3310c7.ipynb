```python
import numpy as np
import numpy.ma as ma
m = ma.masked_equal([0, 1, 0, 2, 3], 0)
m
```

```python
ma.is_mask(m)
```

```python
ma.is_mask(m.mask)
```

Чтобы входные данные считались допустимой маской, они должны быть ndarray или иметь аналогичные атрибуты.


```python
m = [False, True, False]
ma.is_mask(m)
```

```python
m = np.array([False, True, False])
m
```

```python
ma.is_mask(m)
```

Массивы с комплексным dtype не дают True.


```python
dtype = np.dtype({'names':['monty', 'pithon'],
                  'formats':[bool, bool]})
dtype
```

```python
m = np.array([(True, False), (False, True), (True, False)],
             dtype=dtype)
m
```

```python
ma.is_mask(m)
```
