```python
import numpy as np
from numpy.polynomial.polynomial import polyval
polyval(1, [1,2,3])
```

```python
a = np.arange(4).reshape(2,2)
a
```

```python
polyval(a, [1, 2, 3])
```

```python
coef = np.arange(4).reshape(2, 2)  # multidimensional coefficients
coef
```
