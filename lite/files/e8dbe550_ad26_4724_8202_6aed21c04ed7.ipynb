```python
import numpy as np
a = np.arange(4).reshape(2,2)
a
```

```python
a.diagonal()
```

```python
a.diagonal(1)
```

Трёхмерный пример:

```python
a = np.arange(8).reshape(2,2,2); a
```

```python
a.diagonal(0,  # Main diagonals of two arrays created by skipping
           0,  # across the outer(left)-most axis last and
           1)  # the "middle" (row) axis first.
```

**Комментарии к примеру на русском**

Главные диагонали двух массивов, полученных обходом,

при котором крайняя левая, внешняя ось меняется последней, а

средняя ось строк — первой.

Подмассивы, главные диагонали которых мы только что получили. Каждый соответствует фиксированному индексу последней оси (столбцов), а диагонали «упакованы» в строки.


```python
a[:,:,0]  # main diagonal is [0 6]
```

**Комментарии к примеру на русском**

Главная диагональ — [0 6]

```python
a[:,:,1]  # main diagonal is [1 7]
```

**Комментарии к примеру на русском**

Главная диагональ — [1 7]

Антидиагональ может быть получена путем изменения порядка элементов с использованием либо `numpy.flipud`, либо `numpy.fliplr`.

```python
a = np.arange(9).reshape(3, 3)
a
```

```python
np.fliplr(a).diagonal()  # Horizontal flip
```

**Комментарии к примеру на русском**

Отражение по горизонтали

```python
np.flipud(a).diagonal()  # Vertical flip
```

**Комментарии к примеру на русском**

Отражение по вертикали
