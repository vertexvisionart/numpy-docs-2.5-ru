```python
import numpy as np
```

```python
a = np.arange(6).reshape(2,3)
for x in np.nditer(a, flags=['external_loop'], order='F'):
    print(x, end=' ')
```

```python
for x in np.nditer(a, flags=['external_loop','buffered'], order='F'):
    print(x, end=' ')
```
