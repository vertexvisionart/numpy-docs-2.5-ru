```python
import numpy as np
```

Создадим массив 4 × 4.


```python
a = np.arange(16).reshape(4, 4)
a
```

Передайте массив, чтобы получить индексы элементов верхнего треугольника.


```python
triui = np.triu_indices_from(a)
triui
```

```python
a[triui]
```

Это синтаксический сахар для triu_indices().

```python
np.triu_indices(a.shape[0])
```

Используйте параметр `k` для возврата индексов для верхней треугольной матрицы с k-й диагонали.

```python
triuim1 = np.triu_indices_from(a, k=1)
a[triuim1]
```
