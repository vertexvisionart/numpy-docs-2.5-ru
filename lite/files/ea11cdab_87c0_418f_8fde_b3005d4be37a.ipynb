```python
import numpy as np
a = np.arange(9).reshape((3, 3))
m = np.zeros_like(a)
m[1:, 1:] = 1
```

```python
am = np.ma.array(a, mask=m)
np.array(am[~am.mask])
```

```python
np.ma.notmasked_edges(am)
```
