```python
import numpy as np
import numpy.ma as ma
x = ma.masked_equal([0, 1, 0, 2, 3], 0)
x
```

```python
ma.is_masked(x)
```

```python
x = ma.masked_equal([0, 1, 0, 2, 3], 42)
x
```

```python
ma.is_masked(x)
```

Всегда возвращает False, если `x` не является MaskedArray.

```python
x = [False, True, False]
ma.is_masked(x)
```

```python
x = 'a string'
ma.is_masked(x)
```
