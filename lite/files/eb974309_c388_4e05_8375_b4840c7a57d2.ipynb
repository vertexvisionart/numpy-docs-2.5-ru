Преобразование простого dtype:


```python
dt = np.dtype(np.int8)
ctype = np.ctypeslib.as_ctypes_type(dt)
ctype
```

Преобразование структурированного dtype:


```python
dt = np.dtype([('x', 'i4'), ('y', 'f4')])
ctype = np.ctypeslib.as_ctypes_type(dt)
ctype
```
