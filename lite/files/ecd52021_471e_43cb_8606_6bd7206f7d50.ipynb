```python
import numpy as np
np.polyval([3,0,1], 5)  # 3 * 5**2 + 0 * 5**1 + 1
```

```python
np.polyval([3,0,1], np.poly1d(5))
```

```python
np.polyval(np.poly1d([3,0,1]), 5)
```

```python
np.polyval(np.poly1d([3,0,1]), np.poly1d(5))
```
