Создадим массив с замаскированными отрицательными значениями:


```python
import numpy as np
x = np.array([[1, -1, 0], [2, -1, 3], [7, 4, -1]])
masked_x = np.ma.masked_array(x, mask=x < 0)
masked_x
```

Сожмём маскированный массив в одномерный массив немаскированных значений:

```python
np.ma.compressed(masked_x)
```
