2-мерный массив размером 2 х 3, состоящий из 4-байтовых целых элементов:

```python
import numpy as np
```

```python
x = np.array([[1, 2, 3], [4, 5, 6]], np.int32)
type(x)
```

```python
x.shape
```

```python
x.dtype
```

Массив можно индексировать так же, как контейнеры Python:


```python
# The element of x in the *second* row, *third* column, namely, 6.
x[1, 2]
```

**Комментарии к примеру на русском**

Элемент x во *второй* строке и *третьем* столбце, то есть 6.

Например, базовые срезы могут создавать представления массива:


```python
y = x[:,1]
y
```

```python
y[0] = 9 # this also changes the corresponding element in x
y
```
