```python
import numpy as np
```

Построим полином $x^2 + 2x + 3$:


```python
import numpy as np
```

```python
p = np.poly1d([1, 2, 3])
print(np.poly1d(p))
```

Оценить полином в $x = 0.5$:

```python
p(0.5)
```

Найдите корни:

```python
p.r
```

```python
p(p.r)
```

Числа в предыдущей строке равны (0, 0) с машинной точностью.

Выведем коэффициенты:


```python
p.c
```

Выведем порядок (старшие нулевые коэффициенты удаляются):


```python
p.order
```

Покажем коэффициент при k-й степени многочлена. Это эквивалентно ``p.c[-(i+1)]`` при i = k:

```python
p[1]
```

Полиномы можно складывать, вычитать, умножать и делить
(при делении возвращаются частное и остаток):


```python
p * p
```

```python
(p**3 + 4) / p
```

``asarray(p)`` возвращает массив коэффициентов, поэтому многочлены можно передавать во все функции, принимающие массивы:


```python
p**2 # square of polynomial
```

**Комментарии к примеру на русском**

Квадрат многочлена

```python
np.square(p) # square of individual coefficients
```

**Комментарии к примеру на русском**

Квадраты отдельных коэффициентов

Переменную в строковом представлении `p` можно изменить параметром `variable`:


```python
p = np.poly1d([1,2,3], variable='z')
print(p)
```

Постройте полином из его корней:

```python
np.poly1d([1, 2], True)
```

Тот же полином можно получить следующим образом:


```python
np.poly1d([1, -1]) * np.poly1d([1, -2])
```
