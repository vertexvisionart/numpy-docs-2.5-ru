```python
import numpy as np
a = np.array([[1, 2], [3, 4]])
np.mean(a)
```

```python
np.mean(a, axis=0)
```

```python
np.mean(a, axis=1)
```

При одинарной точности результат `mean` может быть неточным:


```python
a = np.zeros((2, 512*512), dtype=np.float32)
a[0, :] = 1.0
a[1, :] = 0.1
np.mean(a)
```

Вычисление среднего значения в float64 является более точным:

```python
np.mean(a, dtype=np.float64)
```

Вычисление среднего доступно и для timedelta64:


```python
b = np.array([1, 3], dtype="timedelta64[D]")
np.mean(b)
```

Указание аргумента where:


```python
a = np.array([[5, 9, 13], [14, 10, 12], [11, 15, 19]])
np.mean(a)
```

```python
np.mean(a, where=[[True], [False], [False]])
```
