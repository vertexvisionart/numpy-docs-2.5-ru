Определяющее свойство первообразной:


```python
import numpy as np
```

```python
p = np.poly1d([1,1,1])
P = np.polyint(p)
P
```

```python
np.polyder(P) == p
```

Константы интегрирования по умолчанию равны нулю, но их можно задать явно:


```python
P = np.polyint(p, 3)
P(0)
```

```python
np.polyder(P)(0)
```

```python
np.polyder(P, 2)(0)
```

```python
P = np.polyint(p, 3, k=[6,5,3])
P
```

Обратите внимание: 3 = 6 / 2!, а константы задаются в порядке интегрирования. Первой указывается константа при члене полинома наивысшей степени:


```python
np.polyder(P, 2)(0)
```

```python
np.polyder(P, 1)(0)
```
