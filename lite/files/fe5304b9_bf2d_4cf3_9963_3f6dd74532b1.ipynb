Число испытаний и вероятность успеха в каждом

Пример из практики. Компания бурит 9 поисковых нефтяных скважин; вероятность успеха для каждой оценивается в 0.1. Все девять скважин оказались неудачными. Какова вероятность такого исхода?

Выполним 20 000 испытаний модели и подсчитаем число испытаний без единого положительного результата.

```python
sum(np.random.binomial(9, 0.1, 20000) == 0)/20000.
```
