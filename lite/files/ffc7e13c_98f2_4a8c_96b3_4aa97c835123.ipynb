```python
import numpy as np
np.searchsorted([11,12,13,14,15], 13)
```

```python
np.searchsorted([11,12,13,14,15], 13, side='right')
```

```python
np.searchsorted([11,12,13,14,15], [-10, 20, 12, 13])
```

При использовании `sorter` возвращаемые индексы относятся к сортированному массиву `a`, а не к самому `a`:

```python
a = np.array([40, 10, 20, 30])
sorter = np.argsort(a)
sorter
```

```python
result = np.searchsorted(a, 25, sorter=sorter)
result
```

```python
a[sorter[result]]
```
