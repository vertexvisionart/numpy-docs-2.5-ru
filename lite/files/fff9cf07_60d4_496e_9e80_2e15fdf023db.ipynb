```python
x.item(7)
```

```python
x.item((0, 1))
```

```python
x.item((2, 2))
```

Для массива с dtype object элементы возвращаются как есть.

```python
a = np.array([np.int64(1)], dtype=np.object_)
a.item() #return np.int64
```
