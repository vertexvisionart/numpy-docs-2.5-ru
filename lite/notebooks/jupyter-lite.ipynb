# jupyter-lite.ipynb

Этот блокнот — предпочтительный источник специфичной для сайта [конфигурации](https://jupyterlite.readthedocs.io/en/latest/configuring.html), используемой приложением JupyterLite во время выполнения. Его настройки переопределяют любую конфигурацию в [`jupyter-lite.json`](./jupyter-lite.json).

## Изменение конфигурации

Конфигурация хранится в метаданных этого блокнота под ключом `jupyter-lite`. Чтобы изменить её в JupyterLab:

- откройте боковую панель _Property Inspector_;
- разверните раздел _Advanced Tools_;
- измените вложенный ключ метаданных `jupyter-lite`;
- нажмите значок с галочкой;
- сохраните блокнот.
